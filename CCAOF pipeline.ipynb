## 2 · Run-flags, diagnostics runner & environment snapshot

`run_stage(...)` wraps every stage: it times it, records status, and on any
exception prints **one copy-paste diagnostic block** and sets an abort flag
(without crashing the notebook) so the summary still runs.

````python
import time, traceback, platform, importlib, json, datetime as _dt

# ---- run flags -------------------------------------------------------------
USE_REAL_DATA  = True    # True = live SEC EDGAR + XBRL + GitHub scrape (uses the firm list below)
FORCE_RESCRAPE = False   # True = ignore the cached raw panel and rebuild from sources
SELECT_MOVERS  = False   # True = restrict to firms that visibly changed CCAI in-window
ALPHA          = 0.05    # significance level for U-test / CIs
RUN_ID = _dt.datetime.now().strftime("%Y-%m-%dT%H:%M:%S")

# ---- shared state ----------------------------------------------------------
S = {"_status": {}, "_timings": {}, "_diagnostics": [], "_aborted": None,
     "run_id": RUN_ID}

def _lib(name):
    try:
        m = importlib.import_module(name)
        return getattr(m, "__version__", "installed")
    except Exception:
        return "MISSING"

def env_str():
    return (f"python={platform.python_version()} | colab={IN_COLAB} | "
            f"pandas={pd.__version__} | numpy={np.__version__} | "
            f"linearmodels={_lib('linearmodels')} | statsmodels={_lib('statsmodels')} | "
            f"sklearn={_lib('sklearn')}")

def _diag_block(stage, err):
    tb = traceback.format_exc()
    bar = "=" * 78
    lines = [
        bar,
        "  CCAOF DIAGNOSTIC  —  copy everything between the bars and paste it back",
        bar,
        f"run_id      : {RUN_ID}",
        f"stage       : {stage}",
        f"status      : FAILED",
        f"error_type  : {type(err).__name__}",
        f"error_msg   : {err}",
        f"environment : {env_str()}",
        f"state_keys  : {[k for k in S if not k.startswith('_')]}",
        "-" * 33 + " TRACEBACK " + "-" * 34,
        tb.rstrip(),
        bar,
    ]
    print("\n".join(lines))

def run_stage(name, fn):
    # Run a stage function; capture timing, status, and clean diagnostics.
    if S["_aborted"]:
        print(f"⏭️  {name}: skipped (a previous stage failed: {S['_aborted']})")
        S["_status"][name] = "skipped"
        return None
    print(f"\n{'─'*78}\n▶ STAGE: {name}\n{'─'*78}")
    t0 = time.time()
    try:
        out = fn()
        dt = time.time() - t0
        S["_status"][name] = "ok"; S["_timings"][name] = dt
        print(f"✅ {name}: OK ({dt:.1f}s)")
        return out
    except Exception as e:
        dt = time.time() - t0
        S["_status"][name] = f"FAILED: {type(e).__name__}"; S["_timings"][name] = dt
        S["_diagnostics"].append((name, traceback.format_exc()))
        S["_aborted"] = name
        _diag_block(name, e)
        return None

def aborted():
    return S["_aborted"] is not None

# ---- helpers to persist frames to Drive ------------------------------------
def save_df(df, path):
    path = Path(path)
    try:
        df.to_parquet(path.with_suffix(".parquet")); return path.with_suffix(".parquet")
    except Exception:
        df.to_csv(path.with_suffix(".csv"), index=False); return path.with_suffix(".csv")

def load_df(path_stem):
    p = Path(path_stem)
    pq, csv = p.with_suffix(".parquet"), p.with_suffix(".csv")
    if pq.exists():
        return pd.read_parquet(pq)
    if csv.exists():
        return pd.read_csv(csv)
    return None

print("Environment health snapshot")
print(" ", env_str())
print(" run_id:", RUN_ID, "| USE_REAL_DATA:", USE_REAL_DATA, "| FORCE_RESCRAPE:", FORCE_RESCRAPE)
````

## 4 · STAGE 1 — Data collection (scrape-once) or synthetic panel

If a cached `panel_raw` exists in Drive and `FORCE_RESCRAPE=False`, it is reused
and collection is skipped. Otherwise: synthetic panel (default), or the real
collectors when `USE_REAL_DATA=True` — **SEC EDGAR 10-K text-mining** (the five
vendor-agnostic CCAI dimensions), **SEC XBRL** (financial controls, official API),
and **year-windowed GitHub** signals (triangulating evidence). No SAP properties,
no scraping, no Yahoo/LinkedIn. The HTTP caches live under Drive so individual
fetches are never repeated across runs.

````python
def _stage_collect():
    cfg = Config.load()
    S["cfg"] = cfg

    # ---- scrape-once short circuit (separate caches for real vs synthetic, so
    # flipping USE_REAL_DATA never reuses the wrong panel) ---------------------
    RAW_NAME = "panel_raw_real" if USE_REAL_DATA else "panel_raw_synth"
    cached = None if FORCE_RESCRAPE else load_df(PATHS["interim"] / RAW_NAME)
    if cached is not None:
        print(f"♻️  Reusing cached raw panel from Drive ({cached.shape}). "
              f"Set FORCE_RESCRAPE=True to rebuild.")
        S["panel"] = cached
        S["data_source"] = "drive-cache"
        return cached

    if USE_REAL_DATA:
        firms = pd.read_csv("data/seed/firms_us200.csv")   # 199 US-listed firms, sector-stratified (edit this CSV to use your own list)
        sess = PoliteSession(cfg.sec_user_agent,
                             rate_per_sec=cfg.rate_limits.get("sec_per_sec", 8))
        edgar = EdgarClient(sess, Cache(PATHS["cache"] / "edgar"))

        # CCAI dimension <- SEC 10-K disclosure-text frequency (firm-year varying).
        DIM_TO_KPI = {
            "standardization": "standardization_kpi",
            "data_harmonization": "data_harmonization_kpi",
            "api_integration": "api_integration_kpi",
            "modular_extensibility": "modular_extensibility_kpi",
            "ops_automation": "ops_automation_kpi",
        }
        krows = []
        cik_map = {}
        for tk in firms["ticker"].dropna():
            cik = edgar.ticker_to_cik(tk)
            if not cik:
                print("  no CIK for", tk); continue
            cik_map[tk] = cik
            for f in edgar.list_filings(cik, ticker=tk, forms=("10-K",),
                                        since_year=min(cfg.years)):
                text = edgar.fetch_filing_text(f)           # cached in Drive
                freqs = DICT.extract_frequencies(text)      # hits / 1000 tokens
                row = {"firm_id": tk, "year": int(f.filing_date[:4])}
                for dim, kpi in DIM_TO_KPI.items():
                    row[kpi] = freqs.get(f"freq_{dim}", 0.0)
                krows.append(row)
        kpis = pd.DataFrame(krows)
        print("  SEC text-mined KPI rows:", kpis.shape)

        # Financial controls from SEC XBRL (official API) — NOT yfinance.
        xbrl = SecXbrlClient(sess, Cache(PATHS["cache"] / "xbrl"))
        controls = xbrl.fetch_controls(cik_map, years=cfg.years)
        print("  XBRL control rows:", controls.shape)

        # Year-windowed GitHub signals (triangulating evidence; org-level counts only).
        gh_rows = []
        if "github_org" in firms.columns:
            gsess = PoliteSession(cfg.sec_user_agent,
                                  rate_per_sec=cfg.rate_limits.get("github_per_min", 25) / 60.0)
            gh = GithubClient(gsess, Cache(PATHS["cache"] / "github"), token=cfg.github_token)
            for _, r in firms.dropna(subset=["github_org"]).iterrows():
                for y in cfg.years:
                    s = gh.engineering_signals(r["github_org"], as_of_year=y)
                    dr = s.dimension_ratios
                    gh_rows.append({"firm_id": r["ticker"], "year": y,
                                    "gh_api_ratio": dr["api_integration"],
                                    "gh_modular_ratio": dr["modular_extensibility"],
                                    "gh_ops_ratio": dr["ops_automation"]})
        gh_df = pd.DataFrame(gh_rows)
        print("  GitHub signal rows:", gh_df.shape)

        # Public OUTCOMES from SEC XBRL (official): operating margin, capitalized-
        # software intensity, SG&A intensity. (time_to_value is not public -> Phase 2.)
        outcomes = xbrl.build_outcomes(cik_map, years=cfg.years)
        print("  XBRL outcome rows:", outcomes.shape)

        # Assemble the firm-year panel from the public inputs.
        panel = kpis.merge(controls, on=["firm_id", "year"], how="left")
        panel = panel.merge(outcomes, on=["firm_id", "year"], how="left")
        if not gh_df.empty:
            panel = panel.merge(gh_df, on=["firm_id", "year"], how="left")

        # A first instrument we CAN build now: leave-one-out industry-year peer CCAI.
        # (Requires an 'industry' column — SIC/NAICS — in your firm sample.)
        if "industry" in firms.columns:
            panel = panel.merge(firms[["ticker", "industry"]]
                                .rename(columns={"ticker": "firm_id"}),
                                on="firm_id", how="left")

        # Build BOTH instruments turnkey:
        #  * z_bartik (primary): firm baseline exposure (low baseline standardization =
        #    more room to move) x national IT-investment shift. Swap the bundled
        #    illustrative shift for your cited FRED series via load_shift_series(fred_series_id=).
        #  * z_peer (over-ID check): leave-one-out industry-year peer CCAI.
        try:
            shift = R.load_shift_series(path="data/seed/it_investment_shift.example.csv",
                                        value="growth")
            panel = R.build_bartik_instrument(panel, shift,
                                              dim_col="standardization_kpi", invert_share=True)
            print("  built z_bartik (replace bundled shift with your FRED series).")
        except Exception as be:
            print("  z_bartik skipped:", be)
        if "industry" in panel.columns:
            panel = R.peer_loo_instrument(panel, group_col="industry")
            print("  built z_peer (leave-one-out industry peer).")

        save_df(panel, PATHS["interim"] / "panel_partial")
        print("  assembled partial panel:", panel.shape, "-> saved as panel_partial")

        # KPI inputs, controls, PUBLIC outcomes, and BOTH instruments are built.
        # Remaining user step: pick the outcome to model and confirm instrument strength.
        # Default modelled outcomes = the public XBRL measures. Drop firm-years with
        # no outcome at all, cache the scrape (one-time), and continue end-to-end.
        before = len(panel)
        panel = panel.dropna(subset=["operating_margin", "capsoft_intensity", "sga_intensity"],
                             how="all").reset_index(drop=True)
        S["data_source"] = "real (SEC EDGAR + XBRL + GitHub)"
        out = save_df(panel, PATHS["interim"] / RAW_NAME)
        print(f"  REAL panel ready: {panel.shape} (dropped {before-len(panel)} rows w/o outcome) -> cached {out}")
        print("  modelled outcomes: operating_margin, capsoft_intensity, sga_intensity")
        print("  NOTE: default sample is the 10-firm example in data/seed/firms.example.csv —")
        print("        replace it with your US-listed firm list for the full study")
        print("        (add an 'industry' column to enable the leave-one-out peer instrument).")
        S["panel"] = panel
        return panel

    # ---- synthetic panel with a planted inverted-U -------------------------
    panel = make_panel(SyntheticConfig(n_firms=200))
    rng = np.random.default_rng(7)
    latent = (panel["ccai"] - panel["ccai"].min()) / (panel["ccai"].max() - panel["ccai"].min())
    for c in ccai.KPI_COLUMNS:
        panel[c] = np.clip(latent + rng.normal(0, 0.12, len(panel)), 0, 1)
    S["data_source"] = "synthetic"

    rep = validate_panel(panel, required=["firm_id", "year", *ccai.KPI_COLUMNS])
    print("  validation:", rep)
    out = save_df(panel, PATHS["interim"] / RAW_NAME)
    print("  saved raw panel ->", out)
    S["panel"] = panel
    return panel

run_stage("data_collection", _stage_collect)
if not aborted():
    display(S["panel"].head())
````

## 5 · STAGE 2 — CCAI construction (0.20 weights, plus PCA later)

Builds the index with the equal **0.20** baseline, reports reliability and
coverage, and saves the panel (with both `ccai_equal` and `ccai_pca`) to Drive. A
distribution chart is written to `visuals/`.

````python
def _stage_index():
    res = ccai.build_ccai(S["panel"])
    S["index_result"] = res
    S["panel_ccai"] = res.panel
    print("equal weights :", res.equal_weights)
    print("PCA weights   :", {k: round(v, 3) for k, v in res.pca_weights.items()})
    print("Cronbach alpha:", round(res.cronbach_alpha, 3),
          "| PCA 1st-eigen share:", round(res.pca_first_eigen_share, 3))
    print("dropped (low coverage):", res.n_dropped_low_coverage)

    # Identifying-variation diagnostic (FE identifies the curve only from firms
    # whose CCAI moves in-window). variation_report summarises it; select_within_change
    # is the lever to focus the sample on movers (set SELECT_MOVERS=True to apply).
    from ccaof.econ import sampling as SMP
    rep = SMP.variation_report(res.panel, ccai_col="ccai_equal")
    S["variation_report"] = rep
    S["within_firm_sd"] = rep["mean_within_sd"]
    print("identifying variation:", {k: (round(v, 3) if isinstance(v, float) else v)
                                      for k, v in rep.items()})
    if rep["share_firms_flat"] > 0.5:
        print("  ⚠️ >50% of firms are flat in CCAI — prioritise firms that changed "
              "in-window (set SELECT_MOVERS=True) and/or add firms.")
    if SELECT_MOVERS:
        before = len(res.panel)
        res.panel = SMP.select_within_change(res.panel, ccai_col="ccai_equal", min_years=3)
        S["panel_ccai"] = res.panel
        print(f"  SELECT_MOVERS: kept movers — {before} → {len(res.panel)} firm-years")

    out = save_df(res.panel, PATHS["processed"] / "panel_ccai")
    print("saved CCAI panel ->", out)

    # distribution chart
    try:
        import matplotlib; matplotlib.use("Agg")
        import matplotlib.pyplot as plt
        fig, ax = plt.subplots(figsize=(7, 4))
        ax.hist(res.panel["ccai_equal"], bins=30, color="#3b6", edgecolor="white")
        ax.set_title("CCAI (equal weights) distribution"); ax.set_xlabel("CCAI"); ax.set_ylabel("firm-years")
        fig.tight_layout()
        p = PATHS["visuals"] / "ccai_distribution.png"; fig.savefig(p, dpi=130)
        S.setdefault("visuals", []).append(str(p)); plt.show(); plt.close(fig)
        print("chart ->", p)
    except Exception as ce:
        print("(chart skipped:", ce, ")")
    return res

run_stage("index_construction", _stage_index)
````

## 6 · STAGE 3 — Quadratic fixed-effects regression + turning point

Fits the two-way FE quadratic for each available outcome (CCAI centered before
squaring), then recovers the turning point with delta + Fieller CIs and the
**Lind–Mehlum U-test**. A "sweet spot" is flagged only when β₁>0, β₂<0, the U-test
passes, **and** the optimum is interior. A fitted-curve chart is saved per outcome.

````python
def _stage_regression():
    df = S["panel_ccai"]
    outcomes = [c for c in ["operating_margin", "capsoft_intensity", "sga_intensity",
                            "tco_proxy", "upgrade_velocity", "time_to_value", "outcome"]
                if c in df.columns]
    controls = [c for c in ["log_revenue", "it_intensity"] if c in df.columns]
    S["outcomes"], S["controls"] = outcomes, controls
    print("outcomes:", outcomes, "| controls:", controls)

    fits = {}
    for y in outcomes:
        fit = P.fit_quadratic_fe(df, outcome=y, controls=controls, ccai_col="ccai_equal")
        lo, hi = P.observed_range(df, fit.center, ccai_col="ccai_equal")
        coef, cov = fit.vertex_inputs()
        tp = TP.recover(coef, cov, lo, hi, alpha=ALPHA, lin_idx=fit.lin_idx, sq_idx=fit.sq_idx)
        b1 = float(fit.params[fit.lin_idx]); b2 = float(fit.params[fit.sq_idx])
        star_orig = (tp.ccai_star + fit.center) if np.isfinite(tp.ccai_star) else float("nan")
        supported = bool(b1 > 0 and b2 < 0 and tp.interior and (tp.u_test_passes or False))
        fits[y] = {"backend": fit.backend, "nobs": fit.nobs, "r2": float(fit.rsquared),
                   "b1": b1, "b2": b2, "ccai_star_orig": star_orig,
                   "interior": bool(tp.interior),
                   "u_test_p": (None if tp.u_test_p is None else float(tp.u_test_p)),
                   "ci_fieller": tp.ci_fieller, "ci_delta": tp.ci_delta,
                   "summary": tp.summary(), "supported": supported,
                   "coef_table": fit.to_frame().round(5).to_dict(),
                   "_fit": fit, "_tp": tp}
        print("=" * 72)
        print(f"OUTCOME: {y}  [backend={fit.backend}, n={fit.nobs}, R2={fit.rsquared:.3f}]")
        display(fit.to_frame().round(4))
        print("  turning point:", tp.summary())
        print("  sweet-spot supported:", supported)

        # fitted within-relationship curve with CCAI* marker + Fieller band
        try:
            import matplotlib; matplotlib.use("Agg")
            import matplotlib.pyplot as plt
            c = np.linspace(lo, hi, 200)
            yhat = b1 * c + b2 * c**2
            x_orig = c + fit.center
            fig, ax = plt.subplots(figsize=(7, 4))
            ax.plot(x_orig, yhat, color="#16c", lw=2, label="fitted quadratic (within)")
            if np.isfinite(star_orig) and lo < (star_orig - fit.center) < hi:
                ax.axvline(star_orig, color="#c33", ls="--", label=f"CCAI* = {star_orig:.3f}")
                if tp.ci_fieller and all(np.isfinite(tp.ci_fieller)):
                    ax.axvspan(tp.ci_fieller[0] + fit.center, tp.ci_fieller[1] + fit.center,
                               color="#c33", alpha=0.10, label="Fieller 95% CI")
            ax.set_title(f"{y}: partial relationship vs CCAI")
            ax.set_xlabel("CCAI (0–1)"); ax.set_ylabel(f"{y} (within-transformed)")
            ax.legend(fontsize=8); fig.tight_layout()
            p = PATHS["visuals"] / f"fit_{y}.png"; fig.savefig(p, dpi=130)
            S.setdefault("visuals", []).append(str(p)); plt.show(); plt.close(fig)
        except Exception as ce:
            print("  (chart skipped:", ce, ")")

    S["fits"] = fits
    return fits

run_stage("regression", _stage_regression)
````

## 7 · STAGE 4 — PCA reweighting (robustness on the index)

Lets the data choose the KPI weights, rebuilds CCAI, re-runs the quadratic on
`ccai_pca`, and reports whether the sweet spot **moves**. A scree chart is saved.

````python
def _stage_pca():
    df = S["panel_ccai"]; res = S["index_result"]
    controls = S["controls"]; outcomes = S["outcomes"]

    w = pd.DataFrame({"equal": res.equal_weights, "pca": res.pca_weights})
    w["gap"] = (w["pca"] - w["equal"]).round(3)
    print("weights (equal vs PCA):"); display(w.round(3))

    # scree
    try:
        from sklearn.decomposition import PCA
        import matplotlib; matplotlib.use("Agg"); import matplotlib.pyplot as plt
        k = res.panel[ccai.KPI_COLUMNS]
        z = (k - k.mean()) / k.std(ddof=0).replace(0, 1)
        ev = PCA().fit(z.fillna(z.median())).explained_variance_ratio_
        fig, ax = plt.subplots(figsize=(6, 4))
        ax.bar([f"PC{i+1}" for i in range(len(ev))], ev * 100, color="#849")
        ax.set_title("Scree — variance explained (%)"); ax.set_ylabel("%")
        fig.tight_layout(); p = PATHS["visuals"] / "scree.png"; fig.savefig(p, dpi=130)
        S.setdefault("visuals", []).append(str(p)); plt.show(); plt.close(fig)
        print("eigen shares:", (ev * 100).round(1))
    except Exception as ce:
        print("(scree skipped:", ce, ")")

    def tp_for(col, y):
        fit = P.fit_quadratic_fe(df, outcome=y, controls=controls, ccai_col=col)
        lo, hi = P.observed_range(df, fit.center, ccai_col=col)
        coef, cov = fit.vertex_inputs()
        tp = TP.recover(coef, cov, lo, hi, alpha=ALPHA, lin_idx=fit.lin_idx, sq_idx=fit.sq_idx)
        return fit.center + tp.ccai_star, bool(tp.interior)

    rows = []
    for y in outcomes:
        se, ei = tp_for("ccai_equal", y)
        sp, pi = tp_for("ccai_pca", y)
        rows.append({"outcome": y, "CCAI*_equal": round(se, 3), "CCAI*_pca": round(sp, 3),
                     "shift": round(sp - se, 3), "equal_interior": ei, "pca_interior": pi})
    cmp = pd.DataFrame(rows)
    print("turning point: equal vs PCA weights"); display(cmp)
    S["pca_compare"] = cmp.to_dict("records")
    return cmp

run_stage("pca_robustness", _stage_pca)
````

## 8 · STAGE 5 — Robustness battery (control function, IV, Driscoll–Kraay, subsample, placebo)

The **control function (2SRI)** is the recommended endogeneity strategy: it needs an
instrument strong for **CCAI itself** (not a separate strong instrument for CCAI²), so
it dissolves the weak squared-instrument problem, and the residual's coefficient is a
direct **endogeneity test**. The 2SLS-with-Z² estimate is retained as a cross-check (watch
the squared-term F — the known weak point). On real data, build the instrument with
`R.shift_share_instrument(...)` (Bartik) and/or `R.peer_loo_instrument(...)` (leave-one-out
peer), then pass its column below. Driscoll–Kraay carries the short-T caveat; subsample
splits the sample; placebo is a scaffold for a non-adopter control panel.

````python
def _stage_robustness():
    df = S["panel_ccai"]; controls = S["controls"]
    outcome = "outcome" if "outcome" in df.columns else S["outcomes"][0]
    rb = {"outcome": outcome}

    # ---- IV (two instruments) ----------------------------------------------
    if "partner_density" in df.columns:
        iv = R.iv_quadratic(df, outcome=outcome, instrument="partner_density",
                            controls=controls, ccai_col="ccai_equal")
        fs = iv.get("first_stage_F", {}); names = iv["names"]
        fvals = list(fs.values())
        F_lin = fs.get(names[0], fvals[0] if len(fvals) > 0 else float("nan"))
        F_sq  = fs.get(names[1], fvals[1] if len(fvals) > 1 else float("nan"))
        lo, hi = P.observed_range(df, iv["center"], ccai_col="ccai_equal")
        tp_iv = TP.recover(iv["params"], iv["cov"], lo, hi, alpha=ALPHA, lin_idx=0, sq_idx=1)
        rb["iv"] = {"backend": iv["backend"], "F_linear": float(F_lin), "F_squared": float(F_sq),
                    "ccai_star_iv": float(iv["center"] + tp_iv.ccai_star),
                    "weak_squared_instrument": bool(F_sq < 10),
                    "summary": tp_iv.summary()}
        print(f"IV backend={iv['backend']}")
        print(f"  first-stage F: linear={F_lin:.1f}  squared={F_sq:.1f}"
              f"  {'⚠️ squared instrument WEAK (<10)' if F_sq < 10 else ''}")
        print(f"  IV CCAI*: {iv['center'] + tp_iv.ccai_star:.3f}  | {tp_iv.summary()}")
    else:
        rb["iv"] = {"note": "no instrument column present"}
        print("IV skipped — no instrument column.")

    # ---- Control function (2SRI) — primary endogeneity strategy -------------
    instr_col = next((c for c in ["z_bartik", "z_peer", "partner_density"]
                      if c in df.columns), None)
    if instr_col is not None:
        cf = R.control_function_quadratic(df, outcome=outcome, instruments=[instr_col],
                                          controls=controls, ccai_col="ccai_equal")
        lo, hi = P.observed_range(df, cf["center"], ccai_col="ccai_equal")
        tp_cf = TP.recover(cf["params"], cf["cov"], lo, hi, alpha=ALPHA,
                           lin_idx=cf["lin_idx"], sq_idx=cf["sq_idx"])
        et = cf["endogeneity_test"]
        rb["control_function"] = {
            "instrument": instr_col, "backend": cf["backend"],
            "first_stage_F": float(cf["first_stage_F"]),
            "weak_instrument": bool(cf["first_stage_F"] < 10),
            "ccai_star_cf": float(cf["center"] + tp_cf.ccai_star),
            "endogeneity_t": float(et["t"]), "endogeneity_p": float(et["p_value"]),
            "reject_exogeneity_5pct": bool(et["reject_exogeneity_5pct"]),
            "summary": tp_cf.summary()}
        print(f"\nControl function (instrument='{instr_col}'):")
        print(f"  first-stage F: {cf['first_stage_F']:.1f}"
              f"  {'⚠️ WEAK (<10)' if cf['first_stage_F'] < 10 else ''}")
        print(f"  endogeneity test: vhat t={et['t']:.2f}, p={et['p_value']:.4f} "
              f"({'endogenous — CF preferred' if et['reject_exogeneity_5pct'] else 'no evidence of endogeneity — FE OK'})")
        print(f"  CF CCAI*: {cf['center'] + tp_cf.ccai_star:.3f}  | {tp_cf.summary()}")
    else:
        rb["control_function"] = {"note": "no instrument column — build one with "
                                  "R.shift_share_instrument / R.peer_loo_instrument"}
        print("\nControl function skipped — construct an instrument column first.")

    # ---- Driscoll–Kraay SEs -------------------------------------------------
    work, cval, lin, sq = P.prepare(df, ccai_col="ccai_equal")
    X = work[[lin, sq] + controls].assign(const=1.0)
    Xv, yv = X.values, work[outcome].values
    beta, *_ = np.linalg.lstsq(Xv, yv, rcond=None)
    resid = yv - Xv @ beta
    dk, dk_lag = R.driscoll_kraay_cov(Xv, resid, work["year"].values)
    se_dk = np.sqrt(np.diag(dk))
    dk_tbl = pd.DataFrame({"coef": np.round(beta, 4), "DK_se": np.round(se_dk, 4)},
                          index=list(X.columns))
    print(f"\nDriscoll–Kraay SEs (Bartlett lag={dk_lag}; short-T caveat applies):")
    display(dk_tbl)
    rb["driscoll_kraay"] = {"bartlett_lag": int(dk_lag),
                            "sq_term_se": float(se_dk[1]), "sq_term_coef": float(beta[1])}

    # ---- subsample stability ------------------------------------------------
    d2 = df.copy()
    grp = next((c for c in ["size_tercile", "industry", "region", "sector"] if c in d2.columns), None)
    if grp is None and "log_revenue" in d2.columns:
        d2["size_tercile"] = pd.qcut(d2["log_revenue"], 3, labels=["S", "M", "L"]); grp = "size_tercile"
    if grp:
        fit_fn = lambda d, outcome, controls: P.fit_quadratic_fe(
            d, outcome=outcome, controls=controls, ccai_col="ccai_equal")
        stab = R.subsample_stability(d2, outcome=outcome, group_col=grp,
                                     fit_fn=fit_fn, controls=controls, min_n=100)
        print(f"\nSubsample stability by {grp}:"); display(stab)
        rb["subsample"] = stab.to_dict("records")
    else:
        rb["subsample"] = {"note": "no grouping column"}

    # ---- placebo (scaffold) -------------------------------------------------
    placebo_df = None   # load a non-SAP control panel here to actually run it
    if placebo_df is not None:
        fit_fn = lambda d, outcome, controls: P.fit_quadratic_fe(
            d, outcome=outcome, controls=controls, ccai_col="ccai_equal")
        rb["placebo"] = R.placebo_test(placebo_df, outcome=outcome, fit_fn=fit_fn, controls=controls)
        print("\nplacebo:", rb["placebo"]["interpretation"])
    else:
        rb["placebo"] = {"note": "no placebo panel supplied (scaffold only)"}
        print("\nplacebo: scaffold only — supply a non-SAP control panel to run it.")

    S["robustness"] = rb
    return rb

run_stage("robustness", _stage_robustness)
````

## 9 · Comprehensive summary  → printed here **and** saved to Drive

This cell **always runs**, even if a stage failed. It assembles findings,
interpretations, robustness, the per-stage status, and any diagnostics into
`SUMMARY.md` + `results.json` in Drive, and prints the whole thing so you can copy
it straight from here.

````python
def _verdict_line(y, f):
    if f["supported"]:
        ci = f.get("ci_fieller")
        ci_s = (f" (Fieller 95% CI [{ci[0]+f['_fit'].center:.3f}, {ci[1]+f['_fit'].center:.3f}])"
                if ci and all(np.isfinite(ci)) else "")
        return (f"- **{y}: sweet spot SUPPORTED.** Optimal CCAI ≈ "
                f"**{f['ccai_star_orig']:.3f}**{ci_s}. β₁={f['b1']:.2f} (>0), "
                f"β₂={f['b2']:.2f} (<0), U-test p={f['u_test_p']:.3f}, interior=yes.")
    reasons = []
    if not (f["b1"] > 0): reasons.append("β₁ not > 0")
    if not (f["b2"] < 0): reasons.append("β₂ not < 0 (no concavity)")
    if not f["interior"]: reasons.append("optimum outside data range")
    if not (f.get("u_test_p") is not None and f["b2"] < 0): pass
    if f.get("u_test_p") is not None and f["b2"] < 0 and not f["supported"] and f["interior"] and f["b1"]>0:
        reasons.append("U-test did not pass")
    return (f"- **{y}: sweet spot NOT supported** ({'; '.join(reasons) or 'criteria not met'}). "
            f"β₁={f['b1']:.2f}, β₂={f['b2']:.2f}. Treat as monotone/null over the observed range — "
            f"a legitimate finding to report as-is.")

def _summary():
    L = []
    A = L.append
    A(f"# CCAOF — Run Summary\n")
    A(f"- **run_id**: {RUN_ID}")
    A(f"- **data source**: {S.get('data_source', 'n/a')}"
      + ("  ⚠️ *synthetic data with a planted optimum — illustrative only*"
         if S.get("data_source") == "synthetic" else ""))
    A(f"- **environment**: {env_str()}")
    A("")

    # stage status table
    A("## Stage status")
    A("| stage | status | seconds |")
    A("|---|---|---|")
    for k in ["imports", "data_collection", "index_construction", "regression",
              "pca_robustness", "robustness"]:
        st = S["_status"].get(k, "not run"); tm = S["_timings"].get(k)
        A(f"| {k} | {st} | {tm:.1f} |" if tm is not None else f"| {k} | {st} | – |")
    A("")

    if S["_aborted"]:
        A(f"> ⛔ **Pipeline aborted at `{S['_aborted']}`.** See the DIAGNOSTICS section "
          f"at the bottom; paste it back to debug.\n")

    # index
    res = S.get("index_result")
    if res is not None:
        A("## Index (CCAI)")
        A(f"- Cronbach α = **{res.cronbach_alpha:.3f}**, PCA 1st-eigenvalue share = "
          f"**{res.pca_first_eigen_share:.3f}**, dropped (low coverage) = {res.n_dropped_low_coverage}.")
        A(f"- mean within-firm SD of CCAI = **{S.get('within_firm_sd', float('nan')):.3f}** "
          f"(must exceed ~0 for fixed-effects identification).")
        A(f"- equal weights {({k: round(v,2) for k,v in res.equal_weights.items()})}")
        A(f"- PCA weights {({k: round(v,2) for k,v in res.pca_weights.items()})}")
        A("")

    # regression verdicts
    fits = S.get("fits")
    if fits:
        A("## Core finding — is there a sweet spot?")
        for y, f in fits.items():
            A(_verdict_line(y, f))
        A("")
        A("### Coefficients")
        for y, f in fits.items():
            A(f"- `{y}`: backend={f['backend']}, n={f['nobs']}, R²={f['r2']:.3f}, "
              f"β₁={f['b1']:.3f}, β₂={f['b2']:.3f}, CCAI*={f['ccai_star_orig']:.3f}.")
        A("")

    # pca compare
    if S.get("pca_compare"):
        A("## Robustness — PCA reweighting")
        for r in S["pca_compare"]:
            tag = "stable" if abs(r["shift"]) <= 0.05 and r["pca_interior"] else "MOVES — weighting matters"
            A(f"- `{r['outcome']}`: CCAI* equal={r['CCAI*_equal']}, PCA={r['CCAI*_pca']}, "
              f"shift={r['shift']} → {tag}.")
        A("")

    # robustness battery
    rb = S.get("robustness")
    if rb:
        A("## Robustness — IV / Driscoll–Kraay / subsample / placebo")
        iv = rb.get("iv", {})
        if "F_linear" in iv:
            A(f"- **IV**: first-stage F linear={iv['F_linear']:.1f}, squared={iv['F_squared']:.1f}"
              + ("  ⚠️ **squared instrument WEAK (<10)** — do not lean on the IV point estimate."
                 if iv.get("weak_squared_instrument") else "")
              + f"  IV CCAI*={iv['ccai_star_iv']:.3f}.")
        else:
            A(f"- **IV**: {iv.get('note','n/a')}")
        dk = rb.get("driscoll_kraay", {})
        if "sq_term_se" in dk:
            A(f"- **Driscoll–Kraay** (lag {dk['bartlett_lag']}): squared-term coef "
              f"{dk['sq_term_coef']:.3f}, DK SE {dk['sq_term_se']:.3f}. Short-T caveat applies.")
        sub = rb.get("subsample")
        if isinstance(sub, list):
            ok = [r for r in sub if r.get("status") == "ok"]
            stars = [f"{r.get('ccai_star'):.3f}" for r in ok if r.get('ccai_star') is not None]
            A(f"- **Subsample**: {len(ok)} groups fit; CCAI* across splits = "
              f"[{', '.join(stars)}] (look for consistency).")
        pl = rb.get("placebo", {})
        A(f"- **Placebo**: {pl.get('interpretation', pl.get('note','n/a'))}")
        A("")

    # standing caveats
    A("## Standing caveats (state these in the dissertation)")
    A("1. **Weak squared-term instrument** is the top fragility — a quadratic needs two "
      "instruments and the squared one is often under-powered; report both first-stage Fs.")
    A("2. **Short panel (~7 years)** limits Driscoll–Kraay reliability and within-firm "
      "identification; lean on two-way clustered SEs for headline inference.")
    A("3. **Construct validity** — text-mining KPIs measure disclosure language, GitHub "
      "signals are noisy org-level proxies; triangulate via the Phase-2 interviews.")
    A("4. **Vendor-agnostic, public sources only** — the index is built from SEC "
      "EDGAR (XBRL + filing text) and GitHub, with no SAP-property data, so there "
      "is no employer conflict of interest. Disclose your SAP affiliation; publish "
      "only with SAP's clearance; keep Phase-2 interviews behind ethics approval.")
    if S.get("data_source") == "synthetic":
        A("5. **These numbers are from synthetic data** with a planted optimum — they "
          "validate the machinery, not the real-world thesis.")
    A("")

    # diagnostics (if any)
    if S["_diagnostics"]:
        A("## DIAGNOSTICS — copy this block to debug")
        for name, tb in S["_diagnostics"]:
            A(f"\n**stage: {name}**\n")
            A("```")
            A(tb.rstrip())
            A("```")

    text = "\n".join(L)

    # write to Drive
    md_path = PATHS["results"] / "SUMMARY.md"
    md_path.write_text(text)
    # machine-readable results
    serial = {
        "run_id": RUN_ID, "data_source": S.get("data_source"),
        "status": S["_status"], "timings": S["_timings"],
        "within_firm_sd": S.get("within_firm_sd"),
        "index": ({"cronbach_alpha": res.cronbach_alpha,
                   "pca_first_eigen_share": res.pca_first_eigen_share,
                   "equal_weights": res.equal_weights, "pca_weights": res.pca_weights}
                  if res is not None else None),
        "fits": ({y: {k: v for k, v in f.items() if not k.startswith("_")}
                  for y, f in (fits or {}).items()}),
        "pca_compare": S.get("pca_compare"),
        "robustness": S.get("robustness"),
        "visuals": S.get("visuals", []),
        "aborted_at": S["_aborted"],
    }
    (PATHS["results"] / "results.json").write_text(json.dumps(serial, indent=2, default=str))

    print(text)
    print("\n" + "=" * 60)
    print("saved:", md_path)
    print("saved:", PATHS["results"] / "results.json")
    if S.get("visuals"):
        print("visuals:", *["\n  " + v for v in S["visuals"]])
    return text

# never wrapped in run_stage — must run even if the pipeline aborted
try:
    _summary()
except Exception as e:
    import traceback as _tb
    print("Summary generation itself failed:\n", _tb.format_exc())
````
