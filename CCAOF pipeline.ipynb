## 2 · Run-flags, diagnostics runner & environment snapshot

`run_stage(...)` wraps every stage: it times it, records status, and on any
exception prints **one copy-paste diagnostic block** and sets an abort flag
(without crashing the notebook) so the summary still runs.

````python
import time, traceback, platform, importlib, json, datetime as _dt

# ---- run flags -------------------------------------------------------------
USE_REAL_DATA  = True    # True = live SEC EDGAR + XBRL + GitHub scrape (uses the firm list below)
FORCE_RESCRAPE = False   # manual override only; leave False — the gate below auto-rebuilds when the cache lacks new columns
SELECT_MOVERS  = False   # True = restrict to firms that visibly changed CCAI in-window
ALPHA          = 0.05    # significance level for U-test / CIs

# ---- enrichment / extra-source flags (additive; safe to leave as default) ---
FETCH_PATENTS     = True   # USPTO PatentsView counts (needs PATENTSVIEW_API_KEY secret); self-skips without a key
FETCH_DEF14A      = True   # SEC DEF 14A IT/cyber-governance text (uses the EDGAR pipeline; no key)
FETCH_TRANSCRIPTS = True   # earnings-call transcripts, ONE call/firm-year (needs FMP_API_KEY); off by default
FETCH_PRICES      = False  # market cap -> Tobin's Q (free no-key prices via stooq.com); off by default
FETCH_GH_RELEASES = False  # GitHub release cadence as an agility outcome; off by default
FETCH_SIMFIN      = True   # SimFin standardized fundamentals + Tobin's Q (needs SIMFIN_API_KEY); self-skips without a key
TRANSCRIPT_PILOT  = True   # if pulling transcripts, restrict to endpoint years x a subsample (cheap robustness layer)
TRANSCRIPT_PILOT_N = 100   # firms in the transcript-pilot subsample
RUN_ID = _dt.datetime.now().strftime("%Y-%m-%dT%H:%M:%S")

# ---- shared state ----------------------------------------------------------
S = {"_status": {}, "_timings": {}, "_diagnostics": [], "_aborted": None,
     "run_id": RUN_ID}

def _lib(name):
    try:
        m = importlib.import_module(name)
        return getattr(m, "__version__", "installed")
    except Exception:
        return "MISSING"

def env_str():
    return (f"python={platform.python_version()} | colab={IN_COLAB} | "
            f"pandas={pd.__version__} | numpy={np.__version__} | "
            f"linearmodels={_lib('linearmodels')} | statsmodels={_lib('statsmodels')} | "
            f"sklearn={_lib('sklearn')}")

def _diag_block(stage, err):
    tb = traceback.format_exc()
    bar = "=" * 78
    lines = [
        bar,
        "  CCAOF DIAGNOSTIC  —  copy everything between the bars and paste it back",
        bar,
        f"run_id      : {RUN_ID}",
        f"stage       : {stage}",
        f"status      : FAILED",
        f"error_type  : {type(err).__name__}",
        f"error_msg   : {err}",
        f"environment : {env_str()}",
        f"state_keys  : {[k for k in S if not k.startswith('_')]}",
        "-" * 33 + " TRACEBACK " + "-" * 34,
        tb.rstrip(),
        bar,
    ]
    print("\n".join(lines))

def run_stage(name, fn):
    # Run a stage function; capture timing, status, and clean diagnostics.
    if S["_aborted"]:
        print(f"⏭️  {name}: skipped (a previous stage failed: {S['_aborted']})")
        S["_status"][name] = "skipped"
        return None
    print(f"\n{'─'*78}\n▶ STAGE: {name}\n{'─'*78}")
    t0 = time.time()
    try:
        out = fn()
        dt = time.time() - t0
        S["_status"][name] = "ok"; S["_timings"][name] = dt
        print(f"✅ {name}: OK ({dt:.1f}s)")
        return out
    except Exception as e:
        dt = time.time() - t0
        S["_status"][name] = f"FAILED: {type(e).__name__}"; S["_timings"][name] = dt
        S["_diagnostics"].append((name, traceback.format_exc()))
        S["_aborted"] = name
        _diag_block(name, e)
        return None

def aborted():
    return S["_aborted"] is not None

# ---- helpers to persist frames to Drive ------------------------------------
def save_df(df, path, retries=4):
    """Persist a frame, tolerating Google Drive FUSE write flakiness (retry)."""
    path = Path(path); last = None
    for attempt in range(retries):
        try:
            path.parent.mkdir(parents=True, exist_ok=True)
            try:
                out = path.with_suffix(".parquet"); df.to_parquet(out)
            except Exception:                       # parquet engine missing -> CSV
                out = path.with_suffix(".csv"); df.to_csv(out, index=False)
            return out
        except (FileNotFoundError, OSError) as err: # transient Drive/FUSE error
            last = err; time.sleep(0.5 * (attempt + 1))
    raise last

def load_df(path_stem):
    p = Path(path_stem)
    pq, csv = p.with_suffix(".parquet"), p.with_suffix(".csv")
    if pq.exists():
        return pd.read_parquet(pq)
    if csv.exists():
        return pd.read_csv(csv)
    return None

print("Environment health snapshot")
print(" ", env_str())
print(" run_id:", RUN_ID, "| USE_REAL_DATA:", USE_REAL_DATA, "| FORCE_RESCRAPE:", FORCE_RESCRAPE)
````

## 4 · STAGE 1 — Data collection (scrape-once) or synthetic panel

If a cached `panel_raw` exists in Drive and `FORCE_RESCRAPE=False`, it is reused
and collection is skipped. Otherwise: synthetic panel (default), or the real
collectors when `USE_REAL_DATA=True` — **SEC EDGAR 10-K text-mining** (the five
vendor-agnostic CCAI dimensions), **SEC XBRL** (financial controls, official API),
and **year-windowed GitHub** signals (triangulating evidence). No SAP properties,
no scraping, no Yahoo/LinkedIn. The HTTP caches live under Drive so individual
fetches are never repeated across runs.

````python
import re as _re
_EMP_RE = _re.compile(
    r"(?:approximately|had|have|employed|employ|total of|were|of)\s+"
    r"(?:approximately\s+)?([0-9][0-9,\.]{1,})\s+"
    r"(?:full[\-\s]?time\s+|regular\s+|salaried\s+)?"
    r"(?:employees|people|persons|team members|associates|staff|workers)",
    _re.IGNORECASE)

def _extract_employees(text):
    """Best-effort headcount from 10-K text (largest plausible mention ~ total)."""
    if not text: return None
    cands = []
    for m in _EMP_RE.finditer(text):
        try: n = int(float(m.group(1).replace(",", "")))
        except Exception: continue
        if 10 <= n <= 3_000_000: cands.append(n)
    return max(cands) if cands else None

def _stage_collect():
    cfg = Config.load()
    S["cfg"] = cfg

    # ---- scrape-once short circuit (separate caches for real vs synthetic, so
    # flipping USE_REAL_DATA never reuses the wrong panel) ---------------------
    RAW_NAME = "panel_raw_real" if USE_REAL_DATA else "panel_raw_synth"
    # Schema marker: columns this build expects in the cached raw panel. If the cache predates a
    # collection upgrade (e.g. before headcount was added) it AUTO-REBUILDS ONCE; the rebuilt
    # cache then contains them, so every later run skips. No manual flag flipping required.
    REQUIRED_RAW_COLS = {"employees"} if USE_REAL_DATA else set()
    cached = None if FORCE_RESCRAPE else load_df(PATHS["interim"] / RAW_NAME)
    if cached is not None and not REQUIRED_RAW_COLS.issubset(set(cached.columns)):
        missing = REQUIRED_RAW_COLS - set(cached.columns)
        print(f"   cached raw panel is missing {sorted(missing)} (schema upgrade) -> rebuilding once to add it.")
        cached = None
    if cached is not None:
        print(f"♻️  Reusing cached raw panel from Drive ({cached.shape}). "
              f"Data present and current -> no scrape. (FORCE_RESCRAPE=True overrides.)")
        S["panel"] = cached
        S["data_source"] = "drive-cache"
        return cached

    print(f"No saved '{RAW_NAME}' yet -> running collection. Per-item fetches read the "
          f"Drive cache when present; finishing this stage saves the panel so future runs skip it.")

    if USE_REAL_DATA:
        firms = pd.read_csv("data/seed/firms_us200.csv")   # 199 US-listed firms, sector-stratified (edit this CSV to use your own list)
        sess = PoliteSession(cfg.sec_user_agent,
                             rate_per_sec=cfg.rate_limits.get("sec_per_sec", 8))
        edgar = EdgarClient(sess, Cache(PATHS["cache"] / "edgar"))

        # CCAI dimension <- SEC 10-K disclosure-text frequency (firm-year varying).
        DIM_TO_KPI = {
            "standardization": "standardization_kpi",
            "data_harmonization": "data_harmonization_kpi",
            "api_integration": "api_integration_kpi",
            "modular_extensibility": "modular_extensibility_kpi",
            "ops_automation": "ops_automation_kpi",
        }
        krows = []
        cik_map = {}
        for tk in firms["ticker"].dropna():
            cik = edgar.ticker_to_cik(tk)
            if not cik:
                print("  no CIK for", tk); continue
            cik_map[tk] = cik
            for f in edgar.list_filings(cik, ticker=tk, forms=("10-K",),
                                        since_year=min(cfg.years)):
                text = edgar.fetch_filing_text(f)           # cached in Drive
                freqs = DICT.extract_frequencies(text)      # hits / 1000 tokens
                row = {"firm_id": tk, "year": int(f.filing_date[:4])}
                for dim, kpi in DIM_TO_KPI.items():
                    row[kpi] = freqs.get(f"freq_{dim}", 0.0)
                row["employees"] = _extract_employees(text)   # headcount -> productivity outcome
                krows.append(row)
        kpis = pd.DataFrame(krows)
        print("  SEC text-mined KPI rows:", kpis.shape)

        # Financial controls from SEC XBRL (official API) — NOT yfinance.
        xbrl = SecXbrlClient(sess, Cache(PATHS["cache"] / "xbrl"))
        controls = xbrl.fetch_controls(cik_map, years=cfg.years)
        print("  XBRL control rows:", controls.shape)

        # Year-windowed GitHub signals (triangulating evidence; org-level counts only).
        gh_rows = []
        if "github_org" in firms.columns:
            gsess = PoliteSession(cfg.sec_user_agent,
                                  rate_per_sec=cfg.rate_limits.get("github_per_min", 25) / 60.0)
            gh = GithubClient(gsess, Cache(PATHS["cache"] / "github"), token=cfg.github_token)
            for _, r in firms.dropna(subset=["github_org"]).iterrows():
                for y in cfg.years:
                    s = gh.engineering_signals(r["github_org"], as_of_year=y)
                    dr = s.dimension_ratios
                    gh_rows.append({"firm_id": r["ticker"], "year": y,
                                    "gh_api_ratio": dr["api_integration"],
                                    "gh_modular_ratio": dr["modular_extensibility"],
                                    "gh_ops_ratio": dr["ops_automation"]})
        gh_df = pd.DataFrame(gh_rows)
        print("  GitHub signal rows:", gh_df.shape)

        # Public OUTCOMES from SEC XBRL (official): operating margin, capitalized-
        # software intensity, SG&A intensity. (time_to_value is not public -> Phase 2.)
        outcomes = xbrl.build_outcomes(cik_map, years=cfg.years)
        print("  XBRL outcome rows:", outcomes.shape)

        # Assemble the firm-year panel from the public inputs.
        panel = kpis.merge(controls, on=["firm_id", "year"], how="left")
        panel = panel.merge(outcomes, on=["firm_id", "year"], how="left")
        # Labour-productivity outcome (free; a more sensitive place for clean-core gains
        # than margin). Noisy because headcount phrasing varies, but it is an OUTCOME so
        # noise only adds variance. NaN where no headcount sentence parsed.
        if "employees" in panel.columns and "revenue" in panel.columns:
            with np.errstate(divide="ignore", invalid="ignore"):
                panel["revenue_per_employee"] = panel["revenue"] / panel["employees"].replace(0, np.nan)
            _n_emp = int(panel["revenue_per_employee"].notna().sum())
            print(f"  headcount parsed -> revenue_per_employee on {_n_emp} firm-years")
        if not gh_df.empty:
            panel = panel.merge(gh_df, on=["firm_id", "year"], how="left")

        # A first instrument we CAN build now: leave-one-out industry-year peer CCAI.
        # (Requires an 'industry' column — SIC/NAICS — in your firm sample.)
        if "industry" in firms.columns:
            panel = panel.merge(firms[["ticker", "industry"]]
                                .rename(columns={"ticker": "firm_id"}),
                                on="firm_id", how="left")

        # Build BOTH instruments turnkey:
        #  * z_bartik (primary): firm baseline exposure (low baseline standardization =
        #    more room to move) x national IT-investment shift. Swap the bundled
        #    illustrative shift for your cited FRED series via load_shift_series(fred_series_id=).
        #  * z_peer (over-ID check): leave-one-out industry-year peer CCAI.
        try:
            shift = R.load_shift_series(path="data/seed/it_investment_shift.example.csv",
                                        value="growth")
            panel = R.build_bartik_instrument(panel, shift,
                                              dim_col="standardization_kpi", invert_share=True)
            print("  built z_bartik (replace bundled shift with your FRED series).")
        except Exception as be:
            print("  z_bartik skipped:", be)
        # z_peer (leave-one-out industry-year peer of CCAI) is built in the
        # index stage, once ccai_equal exists. z_bartik is the primary instrument.

        save_df(panel, PATHS["interim"] / "panel_partial")
        print("  assembled partial panel:", panel.shape, "-> saved as panel_partial")

        # KPI inputs, controls, PUBLIC outcomes, and BOTH instruments are built.
        # Remaining user step: pick the outcome to model and confirm instrument strength.
        # Default modelled outcomes = the public XBRL measures. Drop firm-years with
        # no outcome at all, cache the scrape (one-time), and continue end-to-end.
        before = len(panel)
        panel = panel.dropna(subset=["operating_margin", "capsoft_intensity", "sga_intensity"],
                             how="all").reset_index(drop=True)
        S["data_source"] = "real (SEC EDGAR + XBRL + GitHub)"
        out = save_df(panel, PATHS["interim"] / RAW_NAME)
        print(f"  REAL panel ready: {panel.shape} (dropped {before-len(panel)} rows w/o outcome) -> cached {out}")
        print("  modelled outcomes: operating_margin, capsoft_intensity, sga_intensity")
        print("  NOTE: default sample is the 10-firm example in data/seed/firms.example.csv —")
        print("        replace it with your US-listed firm list for the full study")
        print("        (add an 'industry' column to enable the leave-one-out peer instrument).")
        S["panel"] = panel
        return panel

    # ---- synthetic panel with a planted inverted-U -------------------------
    panel = make_panel(SyntheticConfig(n_firms=200))
    rng = np.random.default_rng(7)
    latent = (panel["ccai"] - panel["ccai"].min()) / (panel["ccai"].max() - panel["ccai"].min())
    for c in ccai.KPI_COLUMNS:
        panel[c] = np.clip(latent + rng.normal(0, 0.12, len(panel)), 0, 1)
    S["data_source"] = "synthetic"

    rep = validate_panel(panel, required=["firm_id", "year", *ccai.KPI_COLUMNS])
    print("  validation:", rep)
    out = save_df(panel, PATHS["interim"] / RAW_NAME)
    print("  saved raw panel ->", out)
    S["panel"] = panel
    return panel

run_stage("data_collection", _stage_collect)
if not aborted():
    display(S["panel"].head())
````

## 5 · STAGE 2 — CCAI construction (0.20 weights, plus PCA later)

Builds the index with the equal **0.20** baseline, reports reliability and
coverage, and saves the panel (with both `ccai_equal` and `ccai_pca`) to Drive. A
distribution chart is written to `visuals/`.

````python
def _stage_index():
    res = ccai.build_ccai(S["panel"])
    S["index_result"] = res
    S["panel_ccai"] = res.panel

    # Leave-one-out industry-year peer instrument, from the actual CCAI
    # (ccai_equal) now that it exists. Over-identifying instrument vs. z_bartik.
    if "industry" in res.panel.columns and "ccai_equal" in res.panel.columns:
        res.panel = R.peer_loo_instrument(res.panel, ccai_col="ccai_equal",
                                          group_col="industry")
        S["panel_ccai"] = res.panel
        print("built z_peer (leave-one-out industry-year peer of CCAI).")
    print("equal weights :", res.equal_weights)
    print("PCA weights   :", {k: round(v, 3) for k, v in res.pca_weights.items()})
    print("Cronbach alpha:", round(res.cronbach_alpha, 3),
          "| PCA 1st-eigen share:", round(res.pca_first_eigen_share, 3))
    print("dropped (low coverage):", res.n_dropped_low_coverage)

    # Identifying-variation diagnostic (FE identifies the curve only from firms
    # whose CCAI moves in-window). variation_report summarises it; select_within_change
    # is the lever to focus the sample on movers (set SELECT_MOVERS=True to apply).
    from ccaof.econ import sampling as SMP
    rep = SMP.variation_report(res.panel, ccai_col="ccai_equal")
    S["variation_report"] = rep
    S["within_firm_sd"] = rep["mean_within_sd"]
    print("identifying variation:", {k: (round(v, 3) if isinstance(v, float) else v)
                                      for k, v in rep.items()})
    if rep["share_firms_flat"] > 0.5:
        print("  ⚠️ >50% of firms are flat in CCAI — prioritise firms that changed "
              "in-window (set SELECT_MOVERS=True) and/or add firms.")
    if SELECT_MOVERS:
        before = len(res.panel)
        res.panel = SMP.select_within_change(res.panel, ccai_col="ccai_equal", min_years=3)
        S["panel_ccai"] = res.panel
        print(f"  SELECT_MOVERS: kept movers — {before} → {len(res.panel)} firm-years")

    # Granular XBRL-derived outcomes (free; finer than operating margin). These
    # flow automatically into regression / cross-firm / dimension because they are
    # just extra columns the candidate-outcome list now looks for.
    p = res.panel
    if "revenue" in p.columns and "assets" in p.columns:
        with np.errstate(divide="ignore", invalid="ignore"):
            p["asset_turnover"] = p["revenue"] / p["assets"].replace(0, np.nan)
            if "operating_margin" in p.columns:
                p["roic_proxy"] = p["operating_margin"] * p["asset_turnover"]
        res.panel = p; S["panel_ccai"] = p
        print("derived granular outcomes:",
              "asset_turnover" + (", roic_proxy" if "operating_margin" in p.columns else ""))

    out = save_df(res.panel, PATHS["processed"] / "panel_ccai")
    print("saved CCAI panel ->", out)

    # distribution chart
    try:
        import matplotlib; matplotlib.use("Agg")
        import matplotlib.pyplot as plt
        fig, ax = plt.subplots(figsize=(7, 4))
        ax.hist(res.panel["ccai_equal"], bins=30, color="#3b6", edgecolor="white")
        ax.set_title("CCAI (equal weights) distribution"); ax.set_xlabel("CCAI"); ax.set_ylabel("firm-years")
        fig.tight_layout()
        p = PATHS["visuals"] / "ccai_distribution.png"; fig.savefig(p, dpi=130)
        S.setdefault("visuals", []).append(str(p)); plt.show(); plt.close(fig)
        print("chart ->", p)
    except Exception as ce:
        print("(chart skipped:", ce, ")")
    return res

run_stage("index_construction", _stage_index)
````

## 6 · STAGE 3 — Quadratic fixed-effects regression + turning point

Fits the two-way FE quadratic for each available outcome (CCAI centered before
squaring), then recovers the turning point with delta + Fieller CIs and the
**Lind–Mehlum U-test**. A "sweet spot" is flagged only when β₁>0, β₂<0, the U-test
passes, **and** the optimum is interior. A fitted-curve chart is saved per outcome.

## SimFin layer — standardized, full-coverage outcomes (+ Tobin's Q)
Runs before regression so the `*_sf` outcomes flow through the whole battery.


````python
# --- Stage: SimFin — standardized fundamentals (outcome-coverage upgrade) + Tobin's Q ---
# Pulls QA'd, standardized financials for the US market via the official `simfin` package
# (bulk download, cached to Drive so it survives past the subscription window — run once
# before it lapses). Adds *_sf outcomes ALONGSIDE the XBRL-derived ones for a direct
# coverage/quality comparison, plus a Tobin's Q proxy from share prices. This is untested
# offline (no key here) — expect one debug pass on the first real run, like SEC/patents.
def _pick(frame, *cands):
    """First matching column: exact, then case-insensitive substring."""
    for c in cands:
        if c in frame.columns: return frame[c]
    low = {col.lower(): col for col in frame.columns}
    for c in cands:
        for lc, col in low.items():
            if c.lower() in lc: return frame[col]
    return None

def _stage_simfin():
    import os
    if not USE_REAL_DATA:
        S["simfin"] = {"note": "skipped (synthetic mode)"}; print("simfin: skipped (synthetic)"); return
    sf_key = os.environ.get("SIMFIN_API_KEY")
    if not (FETCH_SIMFIN and sf_key):
        S["simfin"] = {"note": "skipped (FETCH_SIMFIN off)" if not FETCH_SIMFIN else "skipped (no SIMFIN_API_KEY secret)"}
        print("simfin:", S["simfin"]["note"]); return
    df = S.get("panel_ccai")
    if df is None: S["simfin"] = {"note": "no panel_ccai"}; print("simfin: no panel"); return
    try:
        try:
            import simfin as sf
        except ImportError:
            import subprocess, sys
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", "simfin"], check=False)
            import simfin as sf
        sf.set_api_key(sf_key)
        sdir = PATHS["cache"] / "simfin"; sdir.mkdir(parents=True, exist_ok=True)
        sf.set_data_dir(str(sdir))
        R_DAYS = 365   # use cached bulk download; prevents re-fetch (and free-tier fallback) after the sub lapses
        yrs = set(int(y) for y in pd.Series(df["year"]).dropna().unique())
        want = set(df["firm_id"].astype(str).unique())

        inc = sf.load_income(variant="annual", market="us", refresh_days=R_DAYS).reset_index()
        bal = sf.load_balance(variant="annual", market="us", refresh_days=R_DAYS).reset_index()

        def keyed(fr):
            tk = _pick(fr, "Ticker"); fy = _pick(fr, "Fiscal Year")
            return pd.DataFrame({"firm_id": tk.astype(str), "year": pd.to_numeric(fy, errors="coerce")})

        I = keyed(inc)
        for col, *names in [("revenue_sf", "Revenue"),
                            ("op_income_sf", "Operating Income (Loss)", "Operating Income"),
                            ("gross_profit_sf", "Gross Profit"),
                            ("net_income_sf", "Net Income (Common)", "Net Income"),
                            ("rd_sf", "Research & Development"),
                            ("sga_sf", "Selling, General & Administrative", "SG&A")]:
            v = _pick(inc, *names)
            if v is not None: I[col] = pd.to_numeric(v, errors="coerce")
        B = keyed(bal)
        for col, *names in [("assets_sf", "Total Assets"), ("liab_sf", "Total Liabilities")]:
            v = _pick(bal, *names)
            if v is not None: B[col] = pd.to_numeric(v, errors="coerce")

        m = I.merge(B, on=["firm_id", "year"], how="outer")
        m = m[m["firm_id"].isin(want) & m["year"].isin(yrs)].copy()
        cols = set(m.columns)
        with np.errstate(divide="ignore", invalid="ignore"):
            if {"op_income_sf", "revenue_sf"} <= cols: m["operating_margin_sf"] = m["op_income_sf"] / m["revenue_sf"]
            if {"gross_profit_sf", "revenue_sf"} <= cols: m["gross_margin_sf"] = m["gross_profit_sf"] / m["revenue_sf"]
            if {"net_income_sf", "revenue_sf"} <= cols: m["net_margin_sf"] = m["net_income_sf"] / m["revenue_sf"]
            if {"revenue_sf", "assets_sf"} <= cols: m["asset_turnover_sf"] = m["revenue_sf"] / m["assets_sf"]
            if {"op_income_sf", "assets_sf"} <= cols: m["roic_sf"] = m["op_income_sf"] / m["assets_sf"]
            if {"rd_sf", "revenue_sf"} <= cols: m["rd_intensity_sf"] = m["rd_sf"] / m["revenue_sf"]

        # Tobin's Q — best-effort (share prices are a heavy dataset; fully guarded)
        tq_note = "not attempted"
        try:
            sp = sf.load_shareprices(variant="daily", market="us", refresh_days=R_DAYS).reset_index()
            tkp = _pick(sp, "Ticker"); dt = _pick(sp, "Date"); cl = _pick(sp, "Close")
            sh = _pick(sp, "Shares Outstanding", "Common Shares Outstanding")
            sp2 = pd.DataFrame({"firm_id": tkp.astype(str),
                                "date": pd.to_datetime(dt, errors="coerce"),
                                "close": pd.to_numeric(cl, errors="coerce"),
                                "shares": pd.to_numeric(sh, errors="coerce") if sh is not None else np.nan})
            sp2 = sp2[sp2["firm_id"].isin(want)].dropna(subset=["date"])
            sp2["year"] = sp2["date"].dt.year
            ye = sp2.sort_values("date").groupby(["firm_id", "year"], as_index=False).tail(1)
            ye["mktcap_sf"] = ye["close"] * ye["shares"]
            m = m.merge(ye[["firm_id", "year", "mktcap_sf"]], on=["firm_id", "year"], how="left")
            with np.errstate(divide="ignore", invalid="ignore"):
                if {"mktcap_sf", "liab_sf", "assets_sf"} <= set(m.columns):
                    m["tobin_q"] = (m["mktcap_sf"] + m["liab_sf"]) / m["assets_sf"]
            tq_note = (f"computed on {int(m['tobin_q'].notna().sum())} firm-years"
                       if "tobin_q" in m.columns else "shares/price columns missing")
        except Exception as e:
            tq_note = f"skipped ({type(e).__name__})"

        keep = ["firm_id", "year"] + [c for c in m.columns if c.endswith("_sf") or c == "tobin_q"]
        sf_panel = m[keep].drop_duplicates(["firm_id", "year"])
        df = df.merge(sf_panel, on=["firm_id", "year"], how="left"); S["panel_ccai"] = df
        cov = {c: int(df[c].notna().sum()) for c in keep if c not in ("firm_id", "year")}
        xbrl_cov = int(df["operating_margin"].notna().sum()) if "operating_margin" in df.columns else None
        S["simfin"] = {"coverage": cov, "tobin_q_note": tq_note, "xbrl_opmargin_coverage": xbrl_cov,
                       "summary": (f"merged {len(cov)} standardized columns; operating_margin_sf covers "
                                   f"{cov.get('operating_margin_sf','?')} firm-years"
                                   + (f" vs {xbrl_cov} for the XBRL operating_margin" if xbrl_cov is not None else "")
                                   + f"; Tobin's Q: {tq_note}")}
        print("  simfin:", S["simfin"]["summary"])
        try:
            save_df(df, PATHS["processed"] / "panel_ccai_simfin"); print("  saved panel_ccai_simfin")
        except Exception as se:
            print("  (panel_ccai_simfin save skipped:", se, ")")
    except Exception as e:
        S["simfin"] = {"note": f"failed ({type(e).__name__}: {e})"}; print("  simfin failed:", e)

run_stage("simfin", _stage_simfin)
````

````python
def _stage_regression():
    df = S["panel_ccai"]
    outcomes = [c for c in ["operating_margin", "capsoft_intensity", "sga_intensity",
                            "asset_turnover", "roic_proxy", "revenue_per_employee",
                            "operating_margin_sf", "gross_margin_sf", "net_margin_sf",
                            "asset_turnover_sf", "roic_sf", "rd_intensity_sf", "tobin_q",
                            "tco_proxy", "upgrade_velocity", "time_to_value", "outcome"]
                if c in df.columns]
    controls = [c for c in ["log_revenue", "it_intensity"] if c in df.columns]
    S["outcomes"], S["controls"] = outcomes, controls
    print("outcomes:", outcomes, "| controls:", controls)

    fits = {}
    for y in outcomes:
        fit = P.fit_quadratic_fe(df, outcome=y, controls=controls, ccai_col="ccai_equal")
        lo, hi = P.observed_range(df, fit.center, ccai_col="ccai_equal")
        coef, cov = fit.vertex_inputs()
        tp = TP.recover(coef, cov, lo, hi, alpha=ALPHA, lin_idx=fit.lin_idx, sq_idx=fit.sq_idx)
        b1 = float(fit.params[fit.lin_idx]); b2 = float(fit.params[fit.sq_idx])
        star_orig = (tp.ccai_star + fit.center) if np.isfinite(tp.ccai_star) else float("nan")
        supported = bool(b1 > 0 and b2 < 0 and tp.interior and (tp.u_test_passes or False))
        fits[y] = {"backend": fit.backend, "nobs": fit.nobs, "r2": float(fit.rsquared),
                   "b1": b1, "b2": b2, "ccai_star_orig": star_orig,
                   "interior": bool(tp.interior),
                   "u_test_p": (None if tp.u_test_p is None else float(tp.u_test_p)),
                   "ci_fieller": tp.ci_fieller, "ci_delta": tp.ci_delta,
                   "summary": tp.summary(), "supported": supported,
                   "coef_table": fit.to_frame().round(5).to_dict(),
                   "_fit": fit, "_tp": tp}
        print("=" * 72)
        print(f"OUTCOME: {y}  [backend={fit.backend}, n={fit.nobs}, R2={fit.rsquared:.3f}]")
        display(fit.to_frame().round(4))
        print("  turning point:", tp.summary())
        print("  sweet-spot supported:", supported)

        # fitted within-relationship curve with CCAI* marker + Fieller band
        try:
            import matplotlib; matplotlib.use("Agg")
            import matplotlib.pyplot as plt
            c = np.linspace(lo, hi, 200)
            yhat = b1 * c + b2 * c**2
            x_orig = c + fit.center
            fig, ax = plt.subplots(figsize=(7, 4))
            ax.plot(x_orig, yhat, color="#16c", lw=2, label="fitted quadratic (within)")
            if np.isfinite(star_orig) and lo < (star_orig - fit.center) < hi:
                ax.axvline(star_orig, color="#c33", ls="--", label=f"CCAI* = {star_orig:.3f}")
                if tp.ci_fieller and all(np.isfinite(tp.ci_fieller)):
                    ax.axvspan(tp.ci_fieller[0] + fit.center, tp.ci_fieller[1] + fit.center,
                               color="#c33", alpha=0.10, label="Fieller 95% CI")
            ax.set_title(f"{y}: partial relationship vs CCAI")
            ax.set_xlabel("CCAI (0–1)"); ax.set_ylabel(f"{y} (within-transformed)")
            ax.legend(fontsize=8); fig.tight_layout()
            p = PATHS["visuals"] / f"fit_{y}.png"; fig.savefig(p, dpi=130)
            S.setdefault("visuals", []).append(str(p)); plt.show(); plt.close(fig)
        except Exception as ce:
            print("  (chart skipped:", ce, ")")

    S["fits"] = fits
    return fits

run_stage("regression", _stage_regression)
````

## 7 · STAGE 4 — PCA reweighting (robustness on the index)

Lets the data choose the KPI weights, rebuilds CCAI, re-runs the quadratic on
`ccai_pca`, and reports whether the sweet spot **moves**. A scree chart is saved.

````python
def _stage_pca():
    df = S["panel_ccai"]; res = S["index_result"]
    controls = S["controls"]; outcomes = S["outcomes"]

    w = pd.DataFrame({"equal": res.equal_weights, "pca": res.pca_weights})
    w["gap"] = (w["pca"] - w["equal"]).round(3)
    print("weights (equal vs PCA):"); display(w.round(3))

    # scree
    try:
        from sklearn.decomposition import PCA
        import matplotlib; matplotlib.use("Agg"); import matplotlib.pyplot as plt
        k = res.panel[ccai.KPI_COLUMNS]
        z = (k - k.mean()) / k.std(ddof=0).replace(0, 1)
        ev = PCA().fit(z.fillna(z.median())).explained_variance_ratio_
        fig, ax = plt.subplots(figsize=(6, 4))
        ax.bar([f"PC{i+1}" for i in range(len(ev))], ev * 100, color="#849")
        ax.set_title("Scree — variance explained (%)"); ax.set_ylabel("%")
        fig.tight_layout(); p = PATHS["visuals"] / "scree.png"; fig.savefig(p, dpi=130)
        S.setdefault("visuals", []).append(str(p)); plt.show(); plt.close(fig)
        print("eigen shares:", (ev * 100).round(1))
    except Exception as ce:
        print("(scree skipped:", ce, ")")

    def tp_for(col, y):
        fit = P.fit_quadratic_fe(df, outcome=y, controls=controls, ccai_col=col)
        lo, hi = P.observed_range(df, fit.center, ccai_col=col)
        coef, cov = fit.vertex_inputs()
        tp = TP.recover(coef, cov, lo, hi, alpha=ALPHA, lin_idx=fit.lin_idx, sq_idx=fit.sq_idx)
        return fit.center + tp.ccai_star, bool(tp.interior)

    rows = []
    for y in outcomes:
        se, ei = tp_for("ccai_equal", y)
        sp, pi = tp_for("ccai_pca", y)
        rows.append({"outcome": y, "CCAI*_equal": round(se, 3), "CCAI*_pca": round(sp, 3),
                     "shift": round(sp - se, 3), "equal_interior": ei, "pca_interior": pi})
    cmp = pd.DataFrame(rows)
    print("turning point: equal vs PCA weights"); display(cmp)
    S["pca_compare"] = cmp.to_dict("records")
    return cmp

run_stage("pca_robustness", _stage_pca)
````

## 8 · STAGE 5 — Robustness battery (control function, IV, Driscoll–Kraay, subsample, placebo)

The **control function (2SRI)** is the recommended endogeneity strategy: it needs an
instrument strong for **CCAI itself** (not a separate strong instrument for CCAI²), so
it dissolves the weak squared-instrument problem, and the residual's coefficient is a
direct **endogeneity test**. The 2SLS-with-Z² estimate is retained as a cross-check (watch
the squared-term F — the known weak point). On real data, build the instrument with
`R.shift_share_instrument(...)` (Bartik) and/or `R.peer_loo_instrument(...)` (leave-one-out
peer), then pass its column below. Driscoll–Kraay carries the short-T caveat; subsample
splits the sample; placebo is a scaffold for a non-adopter control panel.

````python
def _stage_robustness():
    df = S["panel_ccai"]; controls = S["controls"]
    outcome = "outcome" if "outcome" in df.columns else S["outcomes"][0]
    rb = {"outcome": outcome}

    # ---- IV (two instruments) ----------------------------------------------
    if "partner_density" in df.columns:
        iv = R.iv_quadratic(df, outcome=outcome, instrument="partner_density",
                            controls=controls, ccai_col="ccai_equal")
        fs = iv.get("first_stage_F", {}); names = iv["names"]
        fvals = list(fs.values())
        F_lin = fs.get(names[0], fvals[0] if len(fvals) > 0 else float("nan"))
        F_sq  = fs.get(names[1], fvals[1] if len(fvals) > 1 else float("nan"))
        lo, hi = P.observed_range(df, iv["center"], ccai_col="ccai_equal")
        tp_iv = TP.recover(iv["params"], iv["cov"], lo, hi, alpha=ALPHA, lin_idx=0, sq_idx=1)
        rb["iv"] = {"backend": iv["backend"], "F_linear": float(F_lin), "F_squared": float(F_sq),
                    "ccai_star_iv": float(iv["center"] + tp_iv.ccai_star),
                    "weak_squared_instrument": bool(F_sq < 10),
                    "summary": tp_iv.summary()}
        print(f"IV backend={iv['backend']}")
        print(f"  first-stage F: linear={F_lin:.1f}  squared={F_sq:.1f}"
              f"  {'⚠️ squared instrument WEAK (<10)' if F_sq < 10 else ''}")
        print(f"  IV CCAI*: {iv['center'] + tp_iv.ccai_star:.3f}  | {tp_iv.summary()}")
    else:
        rb["iv"] = {"note": "no instrument column present"}
        print("IV skipped — no instrument column.")

    # ---- Control function (2SRI) — primary endogeneity strategy -------------
    instr_col = next((c for c in ["z_bartik", "z_peer", "partner_density"]
                      if c in df.columns), None)
    if instr_col is not None:
        cf = R.control_function_quadratic(df, outcome=outcome, instruments=[instr_col],
                                          controls=controls, ccai_col="ccai_equal")
        lo, hi = P.observed_range(df, cf["center"], ccai_col="ccai_equal")
        tp_cf = TP.recover(cf["params"], cf["cov"], lo, hi, alpha=ALPHA,
                           lin_idx=cf["lin_idx"], sq_idx=cf["sq_idx"])
        et = cf["endogeneity_test"]
        rb["control_function"] = {
            "instrument": instr_col, "backend": cf["backend"],
            "first_stage_F": float(cf["first_stage_F"]),
            "weak_instrument": bool(cf["first_stage_F"] < 10),
            "ccai_star_cf": float(cf["center"] + tp_cf.ccai_star),
            "endogeneity_t": float(et["t"]), "endogeneity_p": float(et["p_value"]),
            "reject_exogeneity_5pct": bool(et["reject_exogeneity_5pct"]),
            "summary": tp_cf.summary()}
        print(f"\nControl function (instrument='{instr_col}'):")
        print(f"  first-stage F: {cf['first_stage_F']:.1f}"
              f"  {'⚠️ WEAK (<10)' if cf['first_stage_F'] < 10 else ''}")
        print(f"  endogeneity test: vhat t={et['t']:.2f}, p={et['p_value']:.4f} "
              f"({'endogenous — CF preferred' if et['reject_exogeneity_5pct'] else 'no evidence of endogeneity — FE OK'})")
        print(f"  CF CCAI*: {cf['center'] + tp_cf.ccai_star:.3f}  | {tp_cf.summary()}")
    else:
        rb["control_function"] = {"note": "no instrument column — build one with "
                                  "R.shift_share_instrument / R.peer_loo_instrument"}
        print("\nControl function skipped — construct an instrument column first.")

    # ---- Driscoll–Kraay SEs -------------------------------------------------
    try:
        work, cval, lin, sq = P.prepare(df, ccai_col="ccai_equal")
        X = work[[lin, sq] + controls].assign(const=1.0)
        Xv, yv = X.values, work[outcome].values.astype(float)
        m = np.isfinite(Xv).all(axis=1) & np.isfinite(yv)   # SVD needs finite rows
        Xv, yv, yrs = Xv[m], yv[m], work["year"].values[m]
        if Xv.shape[0] < Xv.shape[1] + 2:
            raise ValueError(f"only {int(Xv.shape[0])} finite rows")
        beta, *_ = np.linalg.lstsq(Xv, yv, rcond=None)
        resid = yv - Xv @ beta
        dk, dk_lag = R.driscoll_kraay_cov(Xv, resid, yrs)
        se_dk = np.sqrt(np.diag(dk))
        dk_tbl = pd.DataFrame({"coef": np.round(beta, 4), "DK_se": np.round(se_dk, 4)},
                              index=list(X.columns))
        print(f"\nDriscoll–Kraay SEs (Bartlett lag={dk_lag}; short-T caveat applies; n={int(Xv.shape[0])}):")
        display(dk_tbl)
        rb["driscoll_kraay"] = {"bartlett_lag": int(dk_lag), "n": int(Xv.shape[0]),
                                "sq_term_se": float(se_dk[1]), "sq_term_coef": float(beta[1])}
    except Exception as dkerr:
        rb["driscoll_kraay"] = {"note": f"skipped ({type(dkerr).__name__}: {dkerr})"}
        print(f"\nDriscoll–Kraay SEs skipped: {dkerr}")

    # ---- subsample stability ------------------------------------------------
    try:
      d2 = df.copy()
      if "ccai_equal" in d2.columns: d2["ccai"] = d2["ccai_equal"]   # alias so observed_range finds the column
      grp = next((c for c in ["size_tercile", "industry", "region", "sector"] if c in d2.columns), None)
      if grp is None and "log_revenue" in d2.columns:
        d2["size_tercile"] = pd.qcut(d2["log_revenue"], 3, labels=["S", "M", "L"]); grp = "size_tercile"
      if grp:
        fit_fn = lambda d, outcome, controls: P.fit_quadratic_fe(
            d, outcome=outcome, controls=controls, ccai_col="ccai_equal")
        stab = R.subsample_stability(d2, outcome=outcome, group_col=grp,
                                     fit_fn=fit_fn, controls=controls, min_n=100)
        print(f"\nSubsample stability by {grp}:"); display(stab)
        rb["subsample"] = stab.to_dict("records")
      else:
        rb["subsample"] = {"note": "no grouping column"}
    except Exception as suberr:
      rb["subsample"] = {"note": f"skipped ({type(suberr).__name__})"}
      print(f"\nSubsample stability skipped: {suberr}")

    # ---- placebo (scaffold) -------------------------------------------------
    placebo_df = None   # load a non-SAP control panel here to actually run it
    if placebo_df is not None:
        fit_fn = lambda d, outcome, controls: P.fit_quadratic_fe(
            d, outcome=outcome, controls=controls, ccai_col="ccai_equal")
        rb["placebo"] = R.placebo_test(placebo_df, outcome=outcome, fit_fn=fit_fn, controls=controls)
        print("\nplacebo:", rb["placebo"]["interpretation"])
    else:
        rb["placebo"] = {"note": "no placebo panel supplied (scaffold only)"}
        print("\nplacebo: scaffold only — supply a non-SAP control panel to run it.")

    S["robustness"] = rb
    return rb

run_stage("robustness", _stage_robustness)
````

## Additional analyses (additive — each stage is self-guarded and never blocks the summary)
Cross-firm (between + Mundlak), construct validity (GitHub convergent + verbosity), richer text features re-mined from the cached 10-Ks, and dimension impact + strategic matrix.

````python
# --- Stage: cross-firm (between + Mundlak) — uses cross-firm variation FE discards ---
def _stage_cross_firm():
    res = {}
    try:
        df = S.get("panel_ccai")
        if df is None or "ccai_equal" not in df.columns:
            S["cross_firm"] = {"note": "no CCAI panel"}; print("cross_firm: skipped (no panel)"); return res
        outs = S.get("outcomes") or []; controls = S.get("controls") or []
        for y in outs:
            sub = df[["firm_id", "ccai_equal", y] + controls].dropna()
            if sub["firm_id"].nunique() < 20:
                res[y] = {"note": f"too few firms ({sub['firm_id'].nunique()})"}; continue
            fm = sub.groupby("firm_id").mean(numeric_only=True)
            c = fm["ccai_equal"].values; ybar = fm[y].values; cc = c - c.mean()
            X = np.column_stack([np.ones_like(cc), cc, cc**2] + [fm[k].values for k in controls])
            beta, *_ = np.linalg.lstsq(X, ybar, rcond=None)
            resid = ybar - X @ beta; n, k = X.shape; s2 = (resid @ resid) / max(n - k, 1)
            cov = s2 * np.linalg.pinv(X.T @ X); lo, hi = float(cc.min()), float(cc.max())
            tp = TP.recover(beta, cov, lo, hi, alpha=ALPHA, lin_idx=1, sq_idx=2)
            b1, b2 = float(beta[1]), float(beta[2])
            star = (tp.ccai_star + c.mean()) if np.isfinite(tp.ccai_star) else float("nan")
            supported = bool(b1 > 0 and b2 < 0 and tp.interior and tp.u_test_passes)
            res[y] = {"n_firms": int(n), "b1": b1, "b2": b2, "ccai_star": star, "interior": bool(tp.interior),
                      "u_test_p": (None if tp.u_test_p is None else float(tp.u_test_p)),
                      "supported": supported, "summary": tp.summary()}
            print(f"  [between] {y}: n_firms={n}, b1={b1:.3f}, b2={b2:.3f}, "
                  f"CCAI*={star:.3f} -> {'INVERTED-U SUPPORTED' if supported else 'not supported'}")
        prim = outs[0] if outs else None
        if prim:
            sub = df[["firm_id", "ccai_equal", prim] + controls].dropna()
            fmean = sub.groupby("firm_id")["ccai_equal"].transform("mean")
            cw = (sub["ccai_equal"] - fmean).values; cb = (fmean - fmean.mean()).values
            X = np.column_stack([np.ones(len(sub)), cw, cw**2, cb, cb**2] + [sub[k].values for k in controls])
            beta, *_ = np.linalg.lstsq(X, sub[prim].values, rcond=None)
            res["_mundlak"] = {"outcome": prim, "within_b1": float(beta[1]), "within_b2": float(beta[2]),
                               "between_b1": float(beta[3]), "between_b2": float(beta[4])}
            print(f"  [mundlak/{prim}] within (b1,b2)=({beta[1]:.3f},{beta[2]:.3f})  "
                  f"between (b1,b2)=({beta[3]:.3f},{beta[4]:.3f})")
    except Exception as e:
        res = {"note": f"failed ({type(e).__name__}: {e})"}; print("  cross_firm error:", e)
    S["cross_firm"] = res; return res

run_stage("cross_firm", _stage_cross_firm)
````

````python
# --- Stage: richer text features (re-mined from cached 10-Ks; adds n_tokens; cached) ---
def _stage_text_features():
    out = {}
    try:
        if not USE_REAL_DATA:
            S["text_rich"] = {"note": "synthetic run - skipped"}; print("text_features: skipped (synthetic)"); return out
        import re as _re
        TR_PATH = PATHS["interim"] / "textrich_panel"
        rich = None if FORCE_RESCRAPE else load_df(TR_PATH)
        if rich is not None:
            print(f"  reusing cached richer-text panel from Drive ({rich.shape}); FORCE_RESCRAPE=True to rebuild.")
        else:
            cfg = S["cfg"]; firms = pd.read_csv("data/seed/firms_us200.csv")
            edgar = EdgarClient(PoliteSession(cfg.sec_user_agent,
                                rate_per_sec=cfg.rate_limits.get("sec_per_sec", 8)),
                                Cache(PATHS["cache"] / "edgar"))
            PHRASES = {
                "standardization": ["clean core", "standard configuration", "out of the box", "reference architecture", "single source of truth"],
                "data_harmonization": ["master data", "data model", "single source of truth", "data harmoniz", "golden record"],
                "api_integration": ["api first", "open api", "rest api", "event driven", "integration layer", "public api"],
                "modular_extensibility": ["side by side", "loosely coupled", "microservice", "modular", "extension framework", "plug in"],
                "ops_automation": ["continuous integration", "continuous delivery", "ci cd", "infrastructure as code", "automated test", "devops"]}
            DIM_TO_KPI = {"standardization": "standardization_kpi", "data_harmonization": "data_harmonization_kpi",
                          "api_integration": "api_integration_kpi", "modular_extensibility": "modular_extensibility_kpi",
                          "ops_automation": "ops_automation_kpi"}
            def _sect(text, starts, ends):
                t = text.lower()
                ss = [m.start() for p in starts for m in [_re.search(p, t)] if m]; s0 = min(ss) if ss else 0
                ee = [m.start() for p in ends for m in [_re.search(p, t)] if m and m.start() > s0]; e0 = min(ee) if ee else len(t)
                return text[s0:e0] if e0 > s0 else text
            rows = []
            for tk in firms["ticker"].dropna():
                cik = edgar.ticker_to_cik(tk)
                if not cik: continue
                for f in edgar.list_filings(cik, ticker=tk, forms=("10-K",), since_year=min(cfg.years)):
                    txt = edgar.fetch_filing_text(f)            # reads Drive cache (no re-download)
                    ntok = len(txt.split())
                    focus = _sect(txt, [r"item\s*7[^a]", r"management.s discussion"], [r"item\s*7a", r"item\s*8"])
                    focus = focus + " " + _sect(txt, [r"item\s*1a", r"risk factors"], [r"item\s*1b", r"item\s*2"])
                    ftl = focus.lower(); flen = max(len(focus.split()), 1)
                    row = {"firm_id": tk, "year": int(f.filing_date[:4]), "n_tokens": ntok}
                    for dim, kpi in DIM_TO_KPI.items():
                        row[kpi] = 1000.0 * sum(ftl.count(p) for p in PHRASES[dim]) / flen
                    rows.append(row)
            rich = pd.DataFrame(rows)
            if rich.empty:
                S["text_rich"] = {"note": "no cached filings found"}; print("text_features: no cached filings"); return out
            save_df(rich, TR_PATH); print(f"  re-mined richer-text panel ({rich.shape}) -> cached to Drive")
        rr = ccai.build_ccai(rich)
        wsd = float(rr.panel.groupby("firm_id")["ccai_equal"].std().mean())
        rp = rr.panel.rename(columns={"ccai_equal": "ccai_textrich"})[["firm_id", "year", "ccai_textrich"]]
        merge_cols = rp.merge(rich[["firm_id", "year", "n_tokens"]], on=["firm_id", "year"])
        base = S.get("panel_ccai")
        if base is not None:
            S["panel_ccai"] = base.merge(merge_cols, on=["firm_id", "year"], how="left")
        comp = S["panel_ccai"][["ccai_equal", "ccai_textrich"]].dropna()
        corr = float(comp["ccai_equal"].corr(comp["ccai_textrich"])) if len(comp) > 5 else float("nan")
        out = {"n_rows": int(len(rich)), "corr_with_baseline": corr,
               "within_firm_sd_textrich": wsd, "cronbach_alpha_textrich": float(rr.cronbach_alpha)}
        print(f"  richer-text CCAI: n={len(rich)}, corr with baseline={corr:.3f}, "
              f"within-firm SD={wsd:.3f} (baseline {S.get('within_firm_sd', float('nan')):.3f}), "
              f"alpha={rr.cronbach_alpha:.3f}")
        prim = (S.get("outcomes") or [None])[0]
        if prim and "ccai_textrich" in S["panel_ccai"].columns:
            sub = S["panel_ccai"][["firm_id", "ccai_textrich", prim]].dropna()
            if sub["firm_id"].nunique() >= 20:
                fm = sub.groupby("firm_id").mean(numeric_only=True)
                c = fm["ccai_textrich"].values - fm["ccai_textrich"].mean()
                X = np.column_stack([np.ones_like(c), c, c**2]); beta, *_ = np.linalg.lstsq(X, fm[prim].values, rcond=None)
                out["between_textrich"] = {"b1": float(beta[1]), "b2": float(beta[2])}
                print(f"  [between/textrich/{prim}] b1={beta[1]:.3f}, b2={beta[2]:.3f}")
    except Exception as e:
        out = {"note": f"failed ({type(e).__name__}: {e})"}; print("  text_features error:", e)
    S["text_rich"] = out; return out

run_stage("text_features", _stage_text_features)
````

````python
# --- Stage: construct validity (GitHub convergent + diagnostic + verbosity control) ---
def _stage_validity():
    out = {}
    try:
        df = S.get("panel_ccai")
        if df is None: S["validity"] = {"note": "no panel"}; print("validity: skipped (no panel)"); return out
        ghcols = [c for c in ["gh_api_ratio", "gh_modular_ratio", "gh_ops_ratio"] if c in df.columns]
        if ghcols:
            d = df.copy(); d["gh_composite"] = d[ghcols].mean(axis=1)
            dd = d[["ccai_equal", "gh_composite"]].dropna()
            ghd = d[ghcols].dropna()
            nf = int(d.dropna(subset=["gh_composite"])["firm_id"].nunique())
            diag = {"n_rows_with_gh": int(len(ghd)),
                    "gh_composite_var": (float(dd["gh_composite"].var()) if len(dd) > 1 else float("nan")),
                    "gh_composite_nunique": (int(dd["gh_composite"].nunique()) if len(dd) else 0),
                    "per_col_var": ({c: float(ghd[c].var()) for c in ghcols} if len(ghd) > 1 else {}),
                    "sample": (ghd.head(5).round(5).to_dict("records") if len(ghd) else [])}
            computable = (len(dd) >= 10 and dd["gh_composite"].nunique() > 1 and dd["ccai_equal"].nunique() > 1)
            if computable:
                pear = float(dd["ccai_equal"].corr(dd["gh_composite"]))
                spear = float(dd["ccai_equal"].corr(dd["gh_composite"], method="spearman"))
                verdict = ("text aligns with revealed GitHub behaviour (convergent)" if (np.isfinite(pear) and pear > 0.2)
                           else "weak/no alignment - disclosure may be decoupled from practice (cheap-talk risk)")
                out["github_convergent"] = {"pearson": pear, "spearman": spear, "n_rows": int(len(dd)),
                                            "n_firms": nf, "verdict": verdict, "diagnostic": diag}
                print(f"  GitHub convergent validity (n_firms={nf}): pearson={pear:.3f}, spearman={spear:.3f} -> {verdict}")
            else:
                out["github_convergent"] = {"n_rows": int(len(dd)), "n_firms": nf,
                    "verdict": "CANNOT COMPUTE - GitHub signal has ~no variance (degenerate); behavioural anchor unusable as collected",
                    "diagnostic": diag}
                print(f"  GitHub convergent validity: CANNOT COMPUTE - gh_composite var={diag['gh_composite_var']:.2e}, "
                      f"nunique={diag['gh_composite_nunique']} (signal essentially constant across {nf} firms)")
                print(f"    per-column variance: {diag['per_col_var']}")
                print(f"    sample gh rows: {diag['sample'][:3]}")
        else:
            out["github_convergent"] = {"note": "no GitHub columns in panel"}
        prim = (S.get("outcomes") or [None])[0]; controls = S.get("controls") or []
        if prim and "n_tokens" in df.columns:
            d = df.copy(); d["log_tokens"] = np.log(d["n_tokens"].clip(lower=1))
            bf = P.fit_quadratic_fe(d, outcome=prim, controls=controls, ccai_col="ccai_equal")
            vf = P.fit_quadratic_fe(d, outcome=prim, controls=controls + ["log_tokens"], ccai_col="ccai_equal")
            b2b = float(bf.params[bf.sq_idx]); b2v = float(vf.params[vf.sq_idx])
            out["verbosity_control"] = {"b2_without": b2b, "b2_with_log_tokens": b2v, "shift": b2v - b2b,
                "verdict": ("robust to disclosure volume" if abs(b2v - b2b) < abs(b2b) * 0.5 + 1e-9
                            else "sensitive to disclosure volume - verbosity confound likely")}
            print(f"  verbosity control ({prim}): beta2 {b2b:.3f} -> {b2v:.3f} after log(10-K tokens)")
        else:
            out["verbosity_control"] = {"note": "n_tokens not in panel (text_features supplies it on a real run)"}
    except Exception as e:
        out = {"note": f"failed ({type(e).__name__}: {e})"}; print("  validity error:", e)
    S["validity"] = out; return out

run_stage("validity", _stage_validity)
````

````python
# --- Stage: dimension/KPI impact + strategic prioritisation matrix ---
def _stage_dimension_impact():
    out = {}
    try:
        df = S.get("panel_ccai"); prim = (S.get("outcomes") or [None])[0]
        if df is None or prim is None:
            S["dimension_impact"] = {"note": "no panel/outcome"}; print("dimension_impact: skipped"); return out
        kpis = [k for k in ["standardization_kpi", "data_harmonization_kpi", "api_integration_kpi",
                            "modular_extensibility_kpi", "ops_automation_kpi"] if k in df.columns]
        sub = df[["firm_id", prim] + kpis].dropna()
        fm = sub.groupby("firm_id").mean(numeric_only=True)
        Z = (fm[kpis] - fm[kpis].mean()) / fm[kpis].std(ddof=0).replace(0, 1)
        X = np.column_stack([np.ones(len(fm))] + [Z[k].values for k in kpis])
        beta, *_ = np.linalg.lstsq(X, fm[prim].values, rcond=None)
        impact = {k: float(b) for k, b in zip(kpis, beta[1:])}; adher = {k: float(fm[k].mean()) for k in kpis}
        med_imp = float(np.median(np.abs(list(impact.values())))); med_lev = float(np.median(list(adher.values())))
        matrix = {}
        for k in kpis:
            hi = abs(impact[k]) >= med_imp; ha = adher[k] >= med_lev
            quad = ("Priority invest (high impact, low adherence)" if hi and not ha else
                    "Maintain (high impact, high adherence)" if hi and ha else
                    "Over-invested (low impact, high adherence)" if not hi and ha else
                    "Deprioritize (low impact, low adherence)")
            matrix[k] = {"impact": impact[k], "adherence_level": adher[k], "quadrant": quad}
        out = {"outcome": prim, "impact": impact, "adherence": adher, "matrix": matrix}
        print("  dimension impact on", prim, "(standardized, between-firm):")
        for k in kpis:
            print(f"    {k:26s} impact={impact[k]:+.3f}  adherence={adher[k]:.3f}  -> {matrix[k]['quadrant']}")
    except Exception as e:
        out = {"note": f"failed ({type(e).__name__}: {e})"}; print("  dimension_impact error:", e)
    S["dimension_impact"] = out; return out

run_stage("dimension_impact", _stage_dimension_impact)
````

## Enrichment & advanced analyses (additive — self-guarded, never block the summary)


````python
# --- Stage: enrichment — extra free sources (patents, DEF 14A; transcripts/prices/releases gated) ---
# Each source is independently guarded: a failure (or a missing key) skips that source
# and leaves the column absent, so downstream stages just don't use it. Network sources
# only run when USE_REAL_DATA. Counts/keys are best-effort and meant to be debugged from a
# live run (paste any DIAGNOSTIC block back).
def _clean_assignee(name):
    s = str(name)
    for suf in [" Inc.", " Inc", " Corporation", " Corp.", " Corp", " Company", " Co.",
                " Co", " Ltd.", " Ltd", " LLC", " plc", " PLC", " Holdings", " Group", ","]:
        s = s.replace(suf, "")
    s = s.replace("The ", "").strip()
    return " ".join(s.split()[:2]) if s else s   # first 1–2 tokens: reduces missed matches

def _stage_enrich():
    import os, time as _t, json as _json
    enrich = {}
    df = S.get("panel_ccai")
    if df is None:
        S["enrich"] = {"note": "no panel_ccai"}; print("enrich: skipped (no panel)"); return enrich
    if not USE_REAL_DATA:
        S["enrich"] = {"note": "skipped (synthetic mode — enrichment fetches real sources only)"}
        print("enrich: skipped (synthetic mode)"); return enrich
    try:
        firms = pd.read_csv("data/seed/firms_us200.csv")
    except Exception as e:
        firms = None; print("  firm list unavailable:", e)
    cfg = S.get("cfg")
    yrs = list(cfg.years) if cfg else sorted(pd.Series(df["year"]).dropna().astype(int).unique())
    lo_y, hi_y = min(yrs), max(yrs)

    # ---------- USPTO patents via PatentsView BULK download (no API key, no account) ----------
    # The PatentSearch API and the USPTO ODP both require a USPTO.gov sign-in now, but PatentsView's
    # bulk TSVs are still openly downloadable from S3 (CC BY 4.0). We pull two tables — granted-patent
    # dates and disambiguated assignees — filter to our firms by organisation-name match, and count
    # grants per firm-year. Heavy (multi-hundred-MB downloads, ~8-10M-row tables) and untested offline;
    # if the S3 layout moved with the ODP migration, set PATENTS_BULK_BASE and re-run. Fully guarded.
    if FETCH_PATENTS and firms is not None:
        try:
            import urllib.request
            base = globals().get("PATENTS_BULK_BASE",
                                  "https://s3.amazonaws.com/data.patentsview.org/download")
            bdir = PATHS["cache"] / "patentsview_bulk"; bdir.mkdir(parents=True, exist_ok=True)
            targets = {"assignee": "g_assignee_disambiguated.tsv.zip", "patent": "g_patent.tsv.zip"}
            paths = {}
            for nm, fn in targets.items():
                lp = bdir / fn
                if not lp.exists() or lp.stat().st_size < 1000:
                    print(f"  patents: downloading {fn} (large, one-time -> Drive cache)...")
                    urllib.request.urlretrieve(f"{base}/{fn}", lp)
                paths[nm] = lp
            # cleaned firm name -> ticker (first 1-2 tokens, lowercased)
            nm_map = {}
            for _, fr in firms.dropna(subset=["company", "ticker"]).iterrows():
                nm_map[_clean_assignee(fr["company"]).lower()] = fr["ticker"]
            wants = [w for w in nm_map if len(w) >= 3]
            keep = []
            for ch in pd.read_csv(paths["assignee"], sep="\t", compression="zip", dtype=str,
                                  usecols=lambda c: c in ("patent_id", "disambig_assignee_organization"),
                                  chunksize=400_000):
                org = ch["disambig_assignee_organization"].fillna("").str.lower()
                hit = pd.Series(False, index=ch.index); who = pd.Series("", index=ch.index)
                for w in wants:
                    m = org.str.contains(w, regex=False) & (~hit)
                    hit = hit | m; who = who.mask(m, nm_map[w])
                if hit.any():
                    sub = ch.loc[hit, ["patent_id"]].copy(); sub["firm_id"] = who[hit].values
                    keep.append(sub)
            amap = pd.concat(keep, ignore_index=True) if keep else pd.DataFrame(columns=["patent_id", "firm_id"])
            if amap.empty:
                raise RuntimeError("no assignee rows matched the firm list (org-name matching)")
            pid = set(amap["patent_id"])
            dts = []
            for ch in pd.read_csv(paths["patent"], sep="\t", compression="zip", dtype=str,
                                  usecols=lambda c: c in ("patent_id", "patent_date"),
                                  chunksize=400_000):
                sub = ch[ch["patent_id"].isin(pid)]
                if not sub.empty: dts.append(sub)
            pdt = pd.concat(dts, ignore_index=True) if dts else pd.DataFrame(columns=["patent_id", "patent_date"])
            merged = amap.merge(pdt, on="patent_id", how="inner")
            merged["year"] = pd.to_numeric(merged["patent_date"].str[:4], errors="coerce")
            merged = merged[merged["year"].isin(set(int(y) for y in yrs))]
            pat = merged.groupby(["firm_id", "year"]).size().reset_index(name="patent_count")
            if not pat.empty:
                pat["year"] = pat["year"].astype(int)
                df = df.merge(pat, on=["firm_id", "year"], how="left")
                if "revenue" in df.columns:
                    df["patent_intensity"] = df["patent_count"] / (df["revenue"].replace(0, np.nan) / 1e9)
                S["panel_ccai"] = df
                enrich["patents"] = {"firm_years": int((pat["patent_count"] > 0).sum()),
                                     "matched_patents": int(len(merged)),
                                     "summary": (f"bulk PatentsView: {int((pat['patent_count']>0).sum())} firm-years "
                                                 f"with >=1 grant (org-name matched, approximate); added patent_count"
                                                 + (", patent_intensity" if "patent_intensity" in df.columns else ""))}
                print("  patents:", enrich["patents"]["summary"])
            else:
                enrich["patents"] = {"note": "matched assignees but no grants in the panel years"}
                print("  patents:", enrich["patents"]["note"])
        except Exception as e:
            enrich["patents"] = {"note": f"failed ({type(e).__name__}: {e}) — if the S3 URLs moved, set PATENTS_BULK_BASE"}
            print("  patents failed:", e)
    else:
        enrich["patents"] = {"note": "skipped (FETCH_PATENTS off)"}
        print("  patents:", enrich["patents"]["note"])
    # ---------- SEC DEF 14A governance/IT-oversight text (existing EDGAR) ----------
    if FETCH_DEF14A and firms is not None:
        try:
            sess = PoliteSession(cfg.sec_user_agent, rate_per_sec=cfg.rate_limits.get("sec_per_sec", 8))
            edg = EdgarClient(sess, Cache(PATHS["cache"] / "edgar"))
            GOV_TERMS = ["cybersecurity", "information security", "data governance", "technology committee",
                         "it oversight", "digital transformation", "chief information officer",
                         "chief technology officer", "cyber risk", "technology risk"]
            grows = []
            sub = firms.dropna(subset=["ticker"])
            for _, fr in sub.iterrows():
                tk = fr["ticker"]; cik = edg.ticker_to_cik(tk)
                if not cik: continue
                for f in edg.list_filings(cik, ticker=tk, forms=("DEF 14A",), since_year=lo_y):
                    try:
                        txt = (edg.fetch_filing_text(f) or "").lower()
                        if not txt: continue
                        toks = max(len(txt.split()), 1)
                        hits = sum(txt.count(t) for t in GOV_TERMS)
                        grows.append({"firm_id": tk, "year": int(f.filing_date[:4]),
                                      "gov_it_signal": 1000.0 * hits / toks})
                    except Exception:
                        continue
            g14 = pd.DataFrame(grows)
            if not g14.empty:
                g14 = g14.groupby(["firm_id", "year"], as_index=False)["gov_it_signal"].mean()
                df = df.merge(g14, on=["firm_id", "year"], how="left"); S["panel_ccai"] = df
                enrich["def14a"] = {"rows": int(len(g14)),
                                    "summary": f"{len(g14)} firm-years of DEF 14A governance text → gov_it_signal "
                                               f"(hits/1k tokens across {len(GOV_TERMS)} IT/cyber-governance terms)"}
                print("  def14a:", enrich["def14a"]["summary"])
            else:
                enrich["def14a"] = {"note": "no DEF 14A text mined"}; print("  def14a:", enrich["def14a"]["note"])
        except Exception as e:
            enrich["def14a"] = {"note": f"failed ({type(e).__name__}: {e})"}; print("  def14a failed:", e)
    else:
        enrich["def14a"] = {"note": "skipped (FETCH_DEF14A off)"}; print("  def14a:", enrich["def14a"]["note"])

    # ---------- earnings-call transcripts (FMP, ONE call/firm-year; off by default) ----------
    fmp_key = os.environ.get("FMP_API_KEY")
    if FETCH_TRANSCRIPTS and fmp_key and firms is not None:
        try:
            import requests
            tc = Cache(PATHS["cache"] / "transcripts")
            trows = []
            t_firms = firms.dropna(subset=["ticker"])
            t_yrs = list(yrs)
            if TRANSCRIPT_PILOT:
                t_yrs = [min(yrs), max(yrs)]
                t_firms = t_firms.head(TRANSCRIPT_PILOT_N)
                print(f"  transcripts: PILOT mode — {len(t_firms)} firms × endpoint years {t_yrs} "
                      f"(~{len(t_firms)*len(t_yrs)} calls)")
            for _, fr in t_firms.iterrows():
                tk = fr["ticker"]
                for y in t_yrs:
                    ckey = f"fmp_{tk}_{y}_q4"
                    j = tc.get_json(ckey)
                    if j is None:
                        try:
                            u = (f"https://financialmodelingprep.com/stable/earning-call-transcript"
                                 f"?symbol={tk}&year={y}&quarter=4&apikey={fmp_key}")
                            r = requests.get(u, timeout=30); r.raise_for_status(); j = r.json()
                            tc.set_json(ckey, j); _t.sleep(0.4)
                        except Exception:
                            j = []
                    content = ""
                    if isinstance(j, list) and j: content = j[0].get("content", "")
                    elif isinstance(j, dict): content = j.get("content", "")
                    if content:
                        freqs = DICT.extract_frequencies(content.lower())
                        vals = [freqs.get(f"freq_{d}", 0.0) for d in
                                ["standardization", "data_harmonization", "api_integration",
                                 "modular_extensibility", "ops_automation"]]
                        trows.append({"firm_id": tk, "year": int(y), "ccai_transcript": float(np.mean(vals))})
            tdf = pd.DataFrame(trows)
            if not tdf.empty:
                df = df.merge(tdf, on=["firm_id", "year"], how="left"); S["panel_ccai"] = df
                enrich["transcripts"] = {"rows": int(len(tdf)),
                                         "summary": f"{len(tdf)} firm-years of Q4 transcript text → ccai_transcript"}
                print("  transcripts:", enrich["transcripts"]["summary"])
            else:
                enrich["transcripts"] = {"note": "no transcript content returned"}
        except Exception as e:
            enrich["transcripts"] = {"note": f"failed ({type(e).__name__}: {e})"}; print("  transcripts failed:", e)
    else:
        enrich["transcripts"] = {"note": ("skipped (FETCH_TRANSCRIPTS off)" if not FETCH_TRANSCRIPTS
                                          else "skipped (no FMP_API_KEY)")}
        print("  transcripts:", enrich["transcripts"]["note"])

    # ---------- Tobin's Q (needs market cap; free no-key prices via stooq.com) ----------
    enrich["tobin_q"] = {"note": ("skipped (FETCH_PRICES off; enable for a forward-looking outcome — "
                                  "free daily prices at stooq.com/q/d/l/?s=TICKER.us, shares from XBRL)")
                         if not FETCH_PRICES else "FETCH_PRICES on but price collector not wired in this build"}
    print("  tobin_q:", enrich["tobin_q"]["note"])

    # ---------- GitHub release cadence (agility outcome; off by default) ----------
    enrich["release_cadence"] = {"note": ("skipped (FETCH_GH_RELEASES off; enable to add release_count/yr as a "
                                          "deployment-velocity outcome for the GitHub subset)")
                                 if not FETCH_GH_RELEASES else "FETCH_GH_RELEASES on but release collector not wired in this build"}
    print("  release_cadence:", enrich["release_cadence"]["note"])

    S["enrich"] = enrich
    try:
        save_df(df, PATHS["processed"] / "panel_ccai_enriched")
        print("  saved enriched panel -> panel_ccai_enriched")
    except Exception as se:
        print("  (enriched-panel save skipped:", se, ")")
    return enrich

run_stage("enrich", _stage_enrich)
````

````python
# --- Stage: advanced — push the data to its limits (horizon, attenuation, granularity) ---
# Self-guarded: every sub-analysis is independently try/except'd and the whole stage
# never raises, so it cannot block the summary. Tackles the three reasons a real effect
# can wash out: (1) horizon mismatch — lagged CCAI + long differences; (2) attenuation —
# disattenuation via index reliability + a second-measure IV; reuses whatever extra
# measure enrichment added (patents/transcripts).
def _stage_advanced():
    out = {}
    try:
        df = S.get("panel_ccai"); prim = (S.get("outcomes") or [None])[0]
        controls = S.get("controls") or []
        if df is None or prim is None or "ccai_equal" not in df.columns:
            S["advanced"] = {"note": "no panel/primary outcome"}; print("advanced: skipped"); return out
        d = df.sort_values(["firm_id", "year"]).copy()

        # ---- HORIZON 1: lagged CCAI (delayed payoffs) ----
        lags = {}
        for L in (1, 2):
            col = f"ccai_lag{L}"
            d[col] = d.groupby("firm_id")["ccai_equal"].shift(L)
            sub = d.dropna(subset=[col, prim] + controls)
            if sub["firm_id"].nunique() >= 20 and len(sub) >= 50:
                try:
                    fit = P.fit_quadratic_fe(sub, outcome=prim, controls=controls, ccai_col=col)
                    lo, hi = P.observed_range(sub, fit.center, ccai_col=col)
                    tp = TP.recover(fit.params, fit.cov, lo, hi, alpha=ALPHA,
                                    lin_idx=fit.lin_idx, sq_idx=fit.sq_idx)
                    lags[f"lag{L}"] = {"b1": float(fit.params[fit.lin_idx]),
                                       "b2": float(fit.params[fit.sq_idx]),
                                       "ccai_star": float(fit.center + tp.ccai_star),
                                       "u_test_passes": bool(tp.u_test_passes), "n": int(len(sub))}
                    print(f"  [lag{L}] {prim}: b1={lags[f'lag{L}']['b1']:.3f}, "
                          f"b2={lags[f'lag{L}']['b2']:.3f}, CCAI*={lags[f'lag{L}']['ccai_star']:.3f}")
                except Exception as e:
                    lags[f"lag{L}"] = {"note": f"fit failed ({type(e).__name__})"}
            else:
                lags[f"lag{L}"] = {"note": "too few firm-years after lagging"}
        out["lagged"] = lags

        # ---- HORIZON 2: long difference (full-window change-on-change) ----
        try:
            g = d.groupby("firm_id")
            ld = pd.DataFrame({"ccai_first": g["ccai_equal"].first(), "ccai_last": g["ccai_equal"].last(),
                               "y_first": g[prim].first(), "y_last": g[prim].last()}).dropna()
            ld["d_ccai"] = ld["ccai_last"] - ld["ccai_first"]; ld["d_y"] = ld["y_last"] - ld["y_first"]
            ld = ld[ld["d_ccai"].abs() > 1e-9]
            if len(ld) >= 20:
                X = np.column_stack([np.ones(len(ld)), ld["d_ccai"].values])
                beta, *_ = np.linalg.lstsq(X, ld["d_y"].values, rcond=None)
                rr = float(ld["d_ccai"].corr(ld["d_y"]))
                out["long_difference"] = {"slope": float(beta[1]), "corr": rr, "n_firms": int(len(ld))}
                print(f"  [long-diff] dCCAI->d{prim}: slope={beta[1]:.3f}, corr={rr:.3f}, n_firms={len(ld)}")
            else:
                out["long_difference"] = {"note": f"too few movers ({len(ld)})"}
        except Exception as e:
            out["long_difference"] = {"note": f"failed ({type(e).__name__})"}

        # ---- ATTENUATION 1: disattenuation via reliability (Cronbach alpha) ----
        try:
            res = S.get("index_result")
            alpha = float(getattr(res, "cronbach_alpha", float("nan"))) if res is not None else float("nan")
            f0 = (S.get("fits") or {}).get(prim)
            if f0 and np.isfinite(alpha) and alpha > 0:
                out["disattenuation"] = {"reliability_alpha": alpha,
                    "b1_obs": f0["b1"], "b2_obs": f0["b2"],
                    "b1_disattenuated": f0["b1"] / alpha, "b2_disattenuated": f0["b2"] / (alpha ** 2),
                    "note": "slope/alpha, curvature/alpha^2; a floor on the true effect since alpha<1"}
                print(f"  [disattenuation] alpha={alpha:.3f}: b1 {f0['b1']:.3f}->{f0['b1']/alpha:.3f}, "
                      f"b2 {f0['b2']:.3f}->{f0['b2']/alpha**2:.3f}")
            else:
                out["disattenuation"] = {"note": "no alpha or primary fit available"}
        except Exception as e:
            out["disattenuation"] = {"note": f"failed ({type(e).__name__})"}

        # ---- ATTENUATION 2: second-measure IV — run for EVERY independent proxy present ----
        try:
            seconds = [c for c in ["patent_intensity", "patent_count", "ccai_transcript"]
                       if c in d.columns and d[c].notna().sum() >= 50 and d[c].nunique() > 5]
            iv = {}
            for second in seconds:
                inst = d[["firm_id", "year", "ccai_equal", prim, second] + controls].dropna()
                if inst["firm_id"].nunique() >= 20 and inst[second].nunique() > 5:
                    try:
                        cf = R.control_function_quadratic(inst, outcome=prim, instruments=[second],
                                                          ccai_col="ccai_equal", controls=controls,
                                                          entity="firm_id", time="year")
                        iv[second] = {"first_stage_F": float(cf["first_stage_F"]),
                                      "b1": float(cf["params"][cf["lin_idx"]]),
                                      "b2": float(cf["params"][cf["sq_idx"]]), "n": int(len(inst))}
                        print(f"  [2nd-measure IV/{second}] first-stage F={iv[second]['first_stage_F']:.1f}, "
                              f"b1={iv[second]['b1']:.3f}, b2={iv[second]['b2']:.3f}")
                    except Exception as e:
                        iv[second] = {"note": f"failed ({type(e).__name__})"}
                else:
                    iv[second] = {"note": f"insufficient variation in {second}"}
            out["second_measure_iv"] = iv if iv else {"note": "no independent second measure present "
                                                              "(enable patents/transcripts in the enrich stage)"}
        except Exception as e:
            out["second_measure_iv"] = {"note": f"failed ({type(e).__name__})"}
    except Exception as e:
        out = {"note": f"failed ({type(e).__name__}: {e})"}; print("  advanced error:", e)
    S["advanced"] = out
    return out

run_stage("advanced", _stage_advanced)
````

````python
def _verdict_line(y, f):
    if f["supported"]:
        ci = f.get("ci_fieller")
        ci_s = (f" (Fieller 95% CI [{ci[0]+f['_fit'].center:.3f}, {ci[1]+f['_fit'].center:.3f}])"
                if ci and all(np.isfinite(ci)) else "")
        return (f"- **{y}: sweet spot SUPPORTED.** Optimal CCAI ≈ "
                f"**{f['ccai_star_orig']:.3f}**{ci_s}. β₁={f['b1']:.2f} (>0), "
                f"β₂={f['b2']:.2f} (<0), U-test p={f['u_test_p']:.3f}, interior=yes.")
    reasons = []
    if not (f["b1"] > 0): reasons.append("β₁ not > 0")
    if not (f["b2"] < 0): reasons.append("β₂ not < 0 (no concavity)")
    if not f["interior"]: reasons.append("optimum outside data range")
    if not (f.get("u_test_p") is not None and f["b2"] < 0): pass
    if f.get("u_test_p") is not None and f["b2"] < 0 and not f["supported"] and f["interior"] and f["b1"]>0:
        reasons.append("U-test did not pass")
    return (f"- **{y}: sweet spot NOT supported** ({'; '.join(reasons) or 'criteria not met'}). "
            f"β₁={f['b1']:.2f}, β₂={f['b2']:.2f}. Treat as monotone/null over the observed range — "
            f"a legitimate finding to report as-is.")

def _summary():
    L = []
    A = L.append
    A(f"# CCAOF — Run Summary\n")
    A(f"- **run_id**: {RUN_ID}")
    A(f"- **data source**: {S.get('data_source', 'n/a')}"
      + ("  ⚠️ *synthetic data with a planted optimum — illustrative only*"
         if S.get("data_source") == "synthetic" else ""))
    A(f"- **environment**: {env_str()}")
    A("")

    # stage status table
    A("## Stage status")
    A("| stage | status | seconds |")
    A("|---|---|---|")
    for k in ["imports", "data_collection", "index_construction", "simfin", "regression",
              "pca_robustness", "robustness", "cross_firm", "text_features",
              "validity", "dimension_impact", "enrich", "advanced"]:
        st = S["_status"].get(k, "not run"); tm = S["_timings"].get(k)
        A(f"| {k} | {st} | {tm:.1f} |" if tm is not None else f"| {k} | {st} | – |")
    A("")

    if S["_aborted"]:
        A(f"> ⛔ **Pipeline aborted at `{S['_aborted']}`.** See the DIAGNOSTICS section "
          f"at the bottom; paste it back to debug.\n")

    # index
    res = S.get("index_result")
    if res is not None:
        A("## Index (CCAI)")
        A(f"- Cronbach α = **{res.cronbach_alpha:.3f}**, PCA 1st-eigenvalue share = "
          f"**{res.pca_first_eigen_share:.3f}**, dropped (low coverage) = {res.n_dropped_low_coverage}.")
        A(f"- mean within-firm SD of CCAI = **{S.get('within_firm_sd', float('nan')):.3f}** "
          f"(must exceed ~0 for fixed-effects identification).")
        A(f"- equal weights {({k: round(v,2) for k,v in res.equal_weights.items()})}")
        A(f"- PCA weights {({k: round(v,2) for k,v in res.pca_weights.items()})}")
        A("")

    # regression verdicts
    fits = S.get("fits")
    if fits:
        A("## Core finding — is there a sweet spot?")
        for y, f in fits.items():
            A(_verdict_line(y, f))
        A("")
        A("### Coefficients")
        for y, f in fits.items():
            A(f"- `{y}`: backend={f['backend']}, n={f['nobs']}, R²={f['r2']:.3f}, "
              f"β₁={f['b1']:.3f}, β₂={f['b2']:.3f}, CCAI*={f['ccai_star_orig']:.3f}.")
        A("")

    # pca compare
    if S.get("pca_compare"):
        A("## Robustness — PCA reweighting")
        for r in S["pca_compare"]:
            tag = "stable" if abs(r["shift"]) <= 0.05 and r["pca_interior"] else "MOVES — weighting matters"
            A(f"- `{r['outcome']}`: CCAI* equal={r['CCAI*_equal']}, PCA={r['CCAI*_pca']}, "
              f"shift={r['shift']} → {tag}.")
        A("")

    # robustness battery
    rb = S.get("robustness")
    if rb:
        A("## Robustness — IV / Driscoll–Kraay / subsample / placebo")
        iv = rb.get("iv", {})
        if "F_linear" in iv:
            A(f"- **IV**: first-stage F linear={iv['F_linear']:.1f}, squared={iv['F_squared']:.1f}"
              + ("  ⚠️ **squared instrument WEAK (<10)** — do not lean on the IV point estimate."
                 if iv.get("weak_squared_instrument") else "")
              + f"  IV CCAI*={iv['ccai_star_iv']:.3f}.")
        else:
            A(f"- **IV**: {iv.get('note','n/a')}")
        dk = rb.get("driscoll_kraay", {})
        if "sq_term_se" in dk:
            A(f"- **Driscoll–Kraay** (lag {dk['bartlett_lag']}): squared-term coef "
              f"{dk['sq_term_coef']:.3f}, DK SE {dk['sq_term_se']:.3f}. Short-T caveat applies.")
        sub = rb.get("subsample")
        if isinstance(sub, list):
            ok = [r for r in sub if r.get("status") == "ok"]
            stars = [f"{r.get('ccai_star'):.3f}" for r in ok if r.get('ccai_star') is not None]
            A(f"- **Subsample**: {len(ok)} groups fit; CCAI* across splits = "
              f"[{', '.join(stars)}] (look for consistency).")
        pl = rb.get("placebo", {})
        A(f"- **Placebo**: {pl.get('interpretation', pl.get('note','n/a'))}")
        A("")

    # cross-firm (between + Mundlak)
    cfd = S.get("cross_firm")
    if cfd and not cfd.get("note"):
        A("## Cross-firm analysis (between + Mundlak) \u2014 does the sweet spot appear across firms?")
        for y, r in cfd.items():
            if y.startswith("_") or not isinstance(r, dict) or "note" in r:
                continue
            tag = "INVERTED-U SUPPORTED across firms" if r.get("supported") else "not supported across firms"
            A(f"- `{y}`: n_firms={r['n_firms']}, \u03b2\u2081={r['b1']:.3f}, \u03b2\u2082={r['b2']:.3f}, "
              f"CCAI*={r['ccai_star']:.3f} \u2192 **{tag}**.")
        mk = cfd.get("_mundlak")
        if mk:
            A(f"- **Mundlak ({mk['outcome']})**: within-firm (\u03b2\u2081={mk['within_b1']:.3f}, "
              f"\u03b2\u2082={mk['within_b2']:.3f}) vs between-firm (\u03b2\u2081={mk['between_b1']:.3f}, "
              f"\u03b2\u2082={mk['between_b2']:.3f}). If between is concave (\u03b2\u2082<0) while within is flat, "
              f"the curve is cross-sectional \u2014 report cross-firm as primary and FE as the "
              f"(underpowered) identification check.")
        A("")

    # construct validity
    vdd = S.get("validity")
    if vdd and not vdd.get("note"):
        A("## Construct validity \u2014 is the text measure real, or cheap talk?")
        gc = vdd.get("github_convergent", {})
        if gc.get("pearson") is not None and np.isfinite(gc.get("pearson", float("nan"))):
            A(f"- **GitHub convergent validity** (n_firms={gc['n_firms']}): Pearson r={gc['pearson']:.3f}, "
              f"Spearman \u03c1={gc['spearman']:.3f} \u2192 {gc['verdict']}.")
        elif "verdict" in gc:
            dgn = gc.get("diagnostic", {})
            A(f"- **GitHub convergent validity**: {gc['verdict']} "
              f"(n_firms={gc.get('n_firms','?')}, gh-composite variance={dgn.get('gh_composite_var', float('nan')):.2e}, "
              f"unique values={dgn.get('gh_composite_nunique','?')}). The revealed-behaviour anchor needs a richer GitHub "
              f"signal or an alternative behavioural source (technographics, hiring, patents) before it can validate the text measure.")
        else:
            A(f"- **GitHub convergent validity**: {gc.get('note', 'n/a')}")
        vc = vdd.get("verbosity_control", {})
        if "b2_with_log_tokens" in vc:
            A(f"- **Verbosity control**: squared-term \u03b2\u2082 {vc['b2_without']:.3f} \u2192 "
              f"{vc['b2_with_log_tokens']:.3f} after adding log(10-K tokens) \u2192 {vc['verdict']}.")
        else:
            A(f"- **Verbosity control**: {vc.get('note', 'n/a')}")
        A("")

    # richer text measure
    trd = S.get("text_rich")
    if trd and not trd.get("note"):
        A("## Richer text measure (section-aware + phrase-level, re-mined from cached 10-Ks)")
        A(f"- n={trd.get('n_rows', '?')} firm-years; correlation with baseline CCAI = "
          f"{trd.get('corr_with_baseline', float('nan')):.3f}; Cronbach \u03b1 = "
          f"{trd.get('cronbach_alpha_textrich', float('nan')):.3f}.")
        A(f"- within-firm SD (richer) = {trd.get('within_firm_sd_textrich', float('nan')):.3f} vs baseline "
          f"{S.get('within_firm_sd', float('nan')):.3f} \u2014 higher is better for FE identification.")
        bt = trd.get("between_textrich")
        if bt:
            A(f"- cross-firm on richer measure: \u03b2\u2081={bt['b1']:.3f}, \u03b2\u2082={bt['b2']:.3f}.")
        A("")

    # dimension/KPI impact + strategic matrix
    did = S.get("dimension_impact")
    if did and not did.get("note"):
        A(f"## Dimension & KPI impact + strategic matrix (outcome: {did.get('outcome', '?')})")
        A("Standardized between-firm association of each adherence dimension with the outcome, "
          "crossed with current adherence level:")
        A("")
        A("| dimension | impact (std) | adherence | quadrant |")
        A("|---|---|---|---|")
        for k, m in did["matrix"].items():
            A(f"| {k} | {m['impact']:+.3f} | {m['adherence_level']:.3f} | {m['quadrant']} |")
        A("")
        A("*Quadrants*: **Priority invest** = high impact / low adherence (act first); **Maintain** = "
          "high impact / high adherence; **Over-invested** = low impact / high adherence; **Deprioritize** "
          "= low impact / low adherence.")
        A("")

    # SimFin standardized outcomes (coverage upgrade)
    sfm = S.get("simfin")
    if sfm and not sfm.get("note"):
        A("## SimFin standardized fundamentals (outcome-coverage upgrade)")
        A(f"- {sfm.get('summary', 'standardized outcomes merged')}")
        cov = sfm.get("coverage") or {}
        if cov:
            A("- Coverage (firm-years): " + ", ".join(f"{k}={v}" for k, v in list(cov.items())[:7]))
        A("")
    elif sfm and sfm.get("note"):
        A(f"_SimFin: {sfm['note']}_"); A("")

    # enrichment (extra free sources)
    enr = S.get("enrich")
    if enr and not enr.get("note"):
        A("## Added data sources (enrichment)")
        for key, label in [("patents", "USPTO patents"), ("def14a", "DEF 14A governance text"),
                           ("transcripts", "earnings-call transcripts"), ("tobin_q", "Tobin's Q"),
                           ("release_cadence", "GitHub release cadence")]:
            r = enr.get(key)
            if not r: continue
            A(f"- **{label}**: {r.get('summary') or r.get('note') or r}")
        A("")

    # advanced (horizon / attenuation / granularity)
    adv = S.get("advanced")
    if adv and not adv.get("note"):
        A("## Pushing the data to its limits \u2014 horizon, attenuation, granularity")
        for k, r in (adv.get("lagged") or {}).items():
            if isinstance(r, dict) and "b1" in r:
                A(f"- **Lagged CCAI ({k})**: \u03b2\u2081={r['b1']:.3f}, \u03b2\u2082={r['b2']:.3f}, "
                  f"CCAI*={r['ccai_star']:.3f}, U-test {'passes' if r['u_test_passes'] else 'fails'} "
                  f"(n={r['n']}) \u2014 does the payoff appear with a delay?")
        ld = adv.get("long_difference", {})
        if "slope" in ld:
            A(f"- **Long difference** (full-window \u0394CCAI\u2192\u0394outcome): slope={ld['slope']:.3f}, "
              f"corr={ld['corr']:.3f}, n_firms={ld['n_firms']} \u2014 change-on-change, robust to fixed levels.")
        da = adv.get("disattenuation", {})
        if "b1_disattenuated" in da:
            A(f"- **Disattenuation** (reliability \u03b1={da['reliability_alpha']:.3f}): "
              f"\u03b2\u2081 {da['b1_obs']:.3f}\u2192{da['b1_disattenuated']:.3f}, "
              f"\u03b2\u2082 {da['b2_obs']:.3f}\u2192{da['b2_disattenuated']:.3f} \u2014 measurement-error-corrected floor.")
        iv2 = adv.get("second_measure_iv", {})
        if iv2.get("note"):
            A(f"- **Second-measure IV**: {iv2['note']}")
        else:
            for meas, r in iv2.items():
                if isinstance(r, dict) and "first_stage_F" in r:
                    wk = "  \u26a0\ufe0f weak instrument (F<10)" if r["first_stage_F"] < 10 else ""
                    A(f"- **Second-measure IV** ({meas}): first-stage F={r['first_stage_F']:.1f}{wk}, "
                      f"\u03b2\u2081={r['b1']:.3f}, \u03b2\u2082={r['b2']:.3f} (n={r['n']}) \u2014 independent proxy instruments the text measure.")
                elif isinstance(r, dict) and r.get("note"):
                    A(f"- **Second-measure IV** ({meas}): {r['note']}")
        A("")

    # standing caveats
    A("## Standing caveats (state these in the dissertation)")
    A("1. **Weak squared-term instrument** is the top fragility — a quadratic needs two "
      "instruments and the squared one is often under-powered; report both first-stage Fs.")
    A("2. **Short panel (~7 years)** limits Driscoll–Kraay reliability and within-firm "
      "identification; lean on two-way clustered SEs for headline inference.")
    A("3. **Construct validity** — text-mining KPIs measure disclosure language, GitHub "
      "signals are noisy org-level proxies; triangulate via the Phase-2 interviews.")
    A("4. **Vendor-agnostic, public sources only** — the index is built from SEC "
      "EDGAR (XBRL + filing text) and GitHub, with no SAP-property data, so there "
      "is no employer conflict of interest. Disclose your SAP affiliation; publish "
      "only with SAP's clearance; keep Phase-2 interviews behind ethics approval.")
    if S.get("data_source") == "synthetic":
        A("5. **These numbers are from synthetic data** with a planted optimum — they "
          "validate the machinery, not the real-world thesis.")
    A("")

    # diagnostics (if any)
    if S["_diagnostics"]:
        A("## DIAGNOSTICS — copy this block to debug")
        for name, tb in S["_diagnostics"]:
            A(f"\n**stage: {name}**\n")
            A("```")
            A(tb.rstrip())
            A("```")

    text = "\n".join(L)

    # write to Drive
    md_path = PATHS["results"] / "SUMMARY.md"
    md_path.write_text(text)
    # machine-readable results
    serial = {
        "run_id": RUN_ID, "data_source": S.get("data_source"),
        "status": S["_status"], "timings": S["_timings"],
        "within_firm_sd": S.get("within_firm_sd"),
        "index": ({"cronbach_alpha": res.cronbach_alpha,
                   "pca_first_eigen_share": res.pca_first_eigen_share,
                   "equal_weights": res.equal_weights, "pca_weights": res.pca_weights}
                  if res is not None else None),
        "fits": ({y: {k: v for k, v in f.items() if not k.startswith("_")}
                  for y, f in (fits or {}).items()}),
        "pca_compare": S.get("pca_compare"),
        "robustness": S.get("robustness"),
        "cross_firm": S.get("cross_firm"),
        "validity": S.get("validity"),
        "text_rich": S.get("text_rich"),
        "dimension_impact": S.get("dimension_impact"),
        "simfin": S.get("simfin"),
        "enrich": S.get("enrich"),
        "advanced": S.get("advanced"),
        "visuals": S.get("visuals", []),
        "aborted_at": S["_aborted"],
    }
    (PATHS["results"] / "results.json").write_text(json.dumps(serial, indent=2, default=str))

    print(text)
    print("\n" + "=" * 60)
    print("saved:", md_path)
    print("saved:", PATHS["results"] / "results.json")
    if S.get("visuals"):
        print("visuals:", *["\n  " + v for v in S["visuals"]])
    return text

# never wrapped in run_stage — must run even if the pipeline aborted
try:
    _summary()
except Exception as e:
    import traceback as _tb
    print("Summary generation itself failed:\n", _tb.format_exc())
````
